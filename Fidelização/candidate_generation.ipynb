```python
import ast
import numpy as np
import pandas as pd

def parse_gt_items_col(test_gt, col="gt_items"):
    """
    Converte test_gt[col] para list[int], lidando com:
    - list/tuple/set
    - numpy arrays
    - strings "[1,2]"
    - NaN
    """
    if col not in test_gt.columns:
        raise ValueError(f"test_gt missing column: {col}")

    def _to_list_int(x):
        # 1) numpy array / pandas array-like
        if isinstance(x, np.ndarray):
            return [int(i) for i in x.tolist()]

        # 2) list/tuple/set
        if isinstance(x, (list, tuple, set)):
            return [int(i) for i in list(x)]

        # 3) NaN / None (tem de vir depois dos casos acima!)
        if x is None:
            return []
        # pd.isna em strings e escalares é ok; em arrays dá erro (já apanhámos acima)
        if isinstance(x, (float, int, np.number)) and pd.isna(x):
            return []

        # 4) string -> literal_eval
        s = str(x).strip()
        if s == "" or s.lower() in {"nan", "none"}:
            return []
        try:
            v = ast.literal_eval(s)
            if isinstance(v, np.ndarray):
                v = v.tolist()
            if isinstance(v, (list, tuple, set)):
                return [int(i) for i in list(v)]
        except Exception:
            pass

        raise ValueError(f"Could not parse gt_items value: {x}")

    test_gt[col] = test_gt[col].apply(_to_list_int)
    return test_gt
```

```python
import ast

def ensure_lower_cols(df):
    df.columns = [c.strip().lower() for c in df.columns]
    return df

def parse_gt_items_col(test_gt, col="gt_items"):
    """
    test_gt['gt_items'] às vezes vem como string "[1,2]".
    Converte para list[int]. Falha cedo se houver lixo.
    """
    if col not in test_gt.columns:
        raise ValueError(f"test_gt missing column: {col}")

    # já é lista?
    sample = test_gt[col].iloc[0]
    if isinstance(sample, list):
        return test_gt

    # string -> list
    def _parse(x):
        if isinstance(x, list):
            return [int(i) for i in x]
        if pd.isna(x):
            return []
        s = str(x).strip()
        try:
            v = ast.literal_eval(s)  # safe parse
            if isinstance(v, (list, tuple)):
                return [int(i) for i in v]
        except Exception:
            pass
        raise ValueError(f"Could not parse gt_items value: {x}")

    test_gt[col] = test_gt[col].apply(_parse)
    return test_gt


# ----------------------- INPUTS -----------------------
# assume que já tens em memória:
# df_cust, df_items, interactions (FULL), baskets (FULL),
# train_interactions, test_gt

# normalize cols
for name in ["df_cust","df_items","interactions","baskets","train_interactions","test_gt"]:
    if name in globals():
        globals()[name] = ensure_lower_cols(globals()[name])

# ---------- enforce minimal schema ----------
# customers
required_cust = {"customer_key", "persona"}
if "df_cust" in globals():
    missing = required_cust - set(df_cust.columns)
    if missing: raise ValueError(f"df_cust missing columns: {missing}")
    df_cust["customer_key"] = df_cust["customer_key"].astype(str)

# items
if "df_items" in globals():
    if "product_id" not in df_items.columns:
        raise ValueError("df_items missing product_id")
    df_items["product_id"] = pd.to_numeric(df_items["product_id"], errors="coerce")
    df_items = df_items.dropna(subset=["product_id"]).copy()
    df_items["product_id"] = df_items["product_id"].astype(int)

# full interactions (opcional)
if "interactions" in globals():
    required_inter = {"customer_key","product_id","strength"}
    missing = required_inter - set(interactions.columns)
    if missing: raise ValueError(f"interactions missing columns: {missing}")
    interactions["customer_key"] = interactions["customer_key"].astype(str)
    interactions["product_id"] = pd.to_numeric(interactions["product_id"], errors="coerce")
    interactions["strength"] = pd.to_numeric(interactions["strength"], errors="coerce").fillna(0.0)
    interactions = interactions.dropna(subset=["product_id"]).copy()
    interactions["product_id"] = interactions["product_id"].astype(int)
    if "last_ts" in interactions.columns:
        interactions["last_ts"] = pd.to_datetime(interactions["last_ts"], errors="coerce")

# baskets (opcional)
if "baskets" in globals():
    required_baskets = {"order_id","product_id"}
    missing = required_baskets - set(baskets.columns)
    if missing: raise ValueError(f"baskets missing columns: {missing}")
    baskets["order_id"] = baskets["order_id"].astype(str)
    baskets["product_id"] = pd.to_numeric(baskets["product_id"], errors="coerce")
    baskets = baskets.dropna(subset=["product_id"]).copy()
    baskets["product_id"] = baskets["product_id"].astype(int)
    if "customer_key" in baskets.columns:
        baskets["customer_key"] = baskets["customer_key"].astype(str)

# train_interactions + test_gt (obrigatórios no novo pipeline)
required_train = {"customer_key","product_id","strength"}
missing = required_train - set(train_interactions.columns)
if missing: raise ValueError(f"train_interactions missing columns: {missing}")

train_interactions["customer_key"] = train_interactions["customer_key"].astype(str)
train_interactions["product_id"] = pd.to_numeric(train_interactions["product_id"], errors="coerce")
train_interactions["strength"] = pd.to_numeric(train_interactions["strength"], errors="coerce").fillna(0.0)
train_interactions = train_interactions.dropna(subset=["product_id"]).copy()
train_interactions["product_id"] = train_interactions["product_id"].astype(int)

if "last_ts" in train_interactions.columns:
    train_interactions["last_ts"] = pd.to_datetime(train_interactions["last_ts"], errors="coerce")

test_gt = parse_gt_items_col(test_gt, "gt_items")
test_gt["customer_key"] = test_gt["customer_key"].astype(str)

# ----------------------- QUALITY GATE -----------------------
print("\n[QC-SPLIT] Basic missing (train/test):")
print("train_interactions:", train_interactions[["customer_key","product_id","strength"]].isna().sum().to_dict())
print("test_gt:", test_gt[["customer_key","gt_items"]].isna().sum().to_dict())

# universos
train_users = set(train_interactions["customer_key"].unique())
train_items = set(train_interactions["product_id"].unique())
test_users  = set(test_gt["customer_key"].unique())

print("\n[QC-SPLIT] Universe sizes:")
print("train users:", len(train_users))
print("train items:", len(train_items))
print("test users:", len(test_users))

# coverage: test users devem estar em df_cust (se existir) e no treino
if "df_cust" in globals():
    cov_test_to_cust = np.mean([u in set(df_cust["customer_key"]) for u in test_users])
    print(f"[QC-SPLIT] Coverage test_gt users -> df_cust: {cov_test_to_cust*100:.2f}%")

cov_test_to_train = np.mean([u in train_users for u in test_users])
print(f"[QC-SPLIT] Coverage test_gt users -> train users: {cov_test_to_train*100:.2f}%")

# sanity: gt_items devem ser listas e não vazias
lens = test_gt["gt_items"].apply(len)
print("\n[QC-SPLIT] gt_items length describe:")
print(lens.describe(percentiles=[.5,.9,.95,.99]))
print("[QC-SPLIT] pct empty gt_items:", (lens==0).mean().round(4))

# leakage check: algum item do test aparece fora do último timestamp? (não dá para 100% sem manteres test_rows)
# mas podemos garantir que train/test não partilham a MESMA interação (user,item,last_ts) se last_ts existir:
if "last_ts" in train_interactions.columns:
    # só dá para check fraco: último timestamp do user não deve estar no treino (para users no teste)
    last_ts_train = train_interactions.groupby("customer_key")["last_ts"].max()
    # aqui assumimos que o test usou o max_ts por user, então o max_ts do treino deve ser < max_ts do teste
    # não temos test_ts, então inferimos pela última compra do gt? impossível. skip.

    pass

# strengths
print("\n[QC] train strength describe:")
print(train_interactions["strength"].describe(percentiles=[.5,.9,.95,.99]))
train_interactions["strength_log"] = np.log1p(train_interactions["strength"])
print("\n[QC] train strength_log describe:")
print(train_interactions["strength_log"].describe(percentiles=[.5,.9,.95,.99]))

# duplicates
dup_ui = train_interactions.duplicated(subset=["customer_key","product_id"]).mean()
print(f"\n[QC] duplicated (customer_key, product_id) in train: {dup_ui*100:.2f}% (ok se agregares depois)")

# optional: items coverage vs df_items
if "df_items" in globals():
    cov_items = np.mean([i in set(df_items["product_id"]) for i in train_items])
    print(f"[QC] Coverage train_items -> df_items: {cov_items*100:.2f}%")

# optional: persona distribution (só para storytelling + sanity)
if "df_cust" in globals():
    print("\n[QC] persona distribution (df_cust):")
    print(df_cust["persona"].value_counts(normalize=True).round(4))
```

## Top Persona

```python
TOP_N_PERSONA = 100  

ti = train_interactions.copy()
ti["customer_key"] = ti["customer_key"].astype(str)
ti["product_id"] = ti["product_id"].astype(int)

# usar log para reduzir skew (alinhado com o QC que já fizeste)
ti["strength"] = pd.to_numeric(ti["strength"], errors="coerce").fillna(0.0)
ti["strength_log"] = np.log1p(ti["strength"])

# juntar persona
cust = df_cust[["customer_key", "persona"]].copy()
cust["customer_key"] = cust["customer_key"].astype(str)

ti = ti.merge(cust, on="customer_key", how="inner")

persona_agg = (
    ti.groupby(["persona", "product_id"])
      .agg(
          score_pop=("strength_log", "sum"),
          n_customers=("customer_key", "nunique"),
          n_events=("customer_key", "size"),
          total_revenue=("strength", "sum"),
      )
      .reset_index()
)

# juntar meta do item 
cols_items = [c for c in ["product_id","brand","hs_code","price_std","price_median","popularity"] if c in df_items_train.columns]
persona_agg = persona_agg.merge(df_items_train[cols_items], on="product_id", how="left")

# ranking dentro de persona
persona_agg = persona_agg.sort_values(
    ["persona", "score_pop", "n_customers", "total_revenue"],
    ascending=[True, False, False, False]
)
persona_agg["rank"] = persona_agg.groupby("persona").cumcount() + 1
persona_top_train = persona_agg[persona_agg["rank"] <= TOP_N_PERSONA].copy()

# QC 
print("[A] persona_top_train:", persona_top_train.shape)
print("[A] unique personas:", persona_top_train["persona"].nunique())
print("[A] coverage items in train:", persona_top_train["product_id"].isin(train_items).mean())

persona_top_train.to_csv(OUT_DIR / "PERSONA_TOP_ITEMS_TRAIN.csv", index=False, encoding="utf-8")
```

## Item Co-occurrence

```python
# --- preparar baskets_train ---
b = baskets.copy()
b["order_id"] = b["order_id"].astype(str)
b["product_id"] = b["product_id"].astype(int)

# manter só itens que aparecem no treino
b = b[b["product_id"].isin(train_items)].copy()

HAS_CUST = "customer_key" in b.columns
HAS_TS   = "created_at" in b.columns

if HAS_CUST:
    b["customer_key"] = b["customer_key"].astype(str)

if HAS_TS:
    b["created_at"] = pd.to_datetime(b["created_at"], errors="coerce")

print("[B] baskets base:", b.shape, "| HAS_CUST:", HAS_CUST, "| HAS_TS:", HAS_TS)

# Tentativa de treino "sem leakage" via split temporal por user (se tiver dados)
if HAS_CUST and HAS_TS:
    # cutoff por user = max last_ts no treino + 1 segundo (garante treino antes do teste)
    # (Como já construíste train_interactions/test_gt, podemos usar o max_ts por user vindo do interactions original
    #  mas aqui fazemos um proxy: cutoff = max last_ts observável no treino do user.)
    cutoff = (
        train_interactions.assign(last_ts=pd.to_datetime(train_interactions["last_ts"], errors="coerce"))
        .groupby("customer_key")["last_ts"].max()
        .rename("cutoff_ts")
        .reset_index()
    )
    b = b.merge(cutoff, on="customer_key", how="inner")
    b_train = b[b["created_at"] <= b["cutoff_ts"]].drop(columns=["cutoff_ts"]).copy()
    print("[B] baskets_train (filtered):", b_train.shape)
else:
    # fallback: não conseguimos filtrar por cliente/tempo
    # mantemos o melhor possível: só items do treino + remove duplicados
    b_train = b.copy()
    print("[B][WARN] baskets sem customer_key/created_at -> cooc pode ter leakage residual. Use com cautela.")
```

```python
TOP_K = 100
MAX_ITEMS_PER_ORDER = 50

def build_item_cooc(baskets_df, min_cooc):
    b = baskets_df[["order_id","product_id"]].dropna().drop_duplicates().copy()

    # cortar orders enormes (evita explosão de pares)
    if MAX_ITEMS_PER_ORDER is not None:
        order_sizes = b.groupby("order_id")["product_id"].nunique()
        keep_orders = order_sizes[order_sizes <= MAX_ITEMS_PER_ORDER].index
        b = b[b["order_id"].isin(keep_orders)].copy()

    n_orders = b["order_id"].nunique()

    # suporte por item (nº de encomendas)
    item_support = (
        b.groupby("product_id")["order_id"].nunique()
         .rename("item_support")
         .reset_index()
    )

    # pares por encomenda
    pairs = b.merge(b, on="order_id", how="inner", suffixes=("_a","_b"))
    pairs = pairs[pairs["product_id_a"] < pairs["product_id_b"]]

    cooc = (
        pairs.groupby(["product_id_a","product_id_b"])["order_id"].nunique()
             .rename("cooc_count")
             .reset_index()
    )

    cooc = cooc[cooc["cooc_count"] >= min_cooc].copy()

    # juntar suportes e scores
    cooc = cooc.merge(item_support, left_on="product_id_a", right_on="product_id", how="left").drop(columns=["product_id"])
    cooc = cooc.rename(columns={"item_support":"support_a"})
    cooc = cooc.merge(item_support, left_on="product_id_b", right_on="product_id", how="left").drop(columns=["product_id"])
    cooc = cooc.rename(columns={"item_support":"support_b"})

    cooc["cosine"]  = cooc["cooc_count"] / np.sqrt(cooc["support_a"] * cooc["support_b"])
    cooc["jaccard"] = cooc["cooc_count"] / (cooc["support_a"] + cooc["support_b"] - cooc["cooc_count"])
    cooc["lift"]    = (cooc["cooc_count"] * n_orders) / (cooc["support_a"] * cooc["support_b"])

    # direcionar e topK
    cooc_ab = cooc.rename(columns={"product_id_a":"product_id","product_id_b":"neighbor_id","support_a":"support_item","support_b":"support_neigh"})
    cooc_ba = cooc.rename(columns={"product_id_b":"product_id","product_id_a":"neighbor_id","support_b":"support_item","support_a":"support_neigh"})
    cooc_dir = pd.concat([cooc_ab, cooc_ba], ignore_index=True)

    SCORE_COL = "cosine"
    cooc_dir = cooc_dir.sort_values(["product_id", SCORE_COL, "cooc_count"], ascending=[True, False, False])
    cooc_dir["rank"] = cooc_dir.groupby("product_id").cumcount() + 1
    topk = cooc_dir[cooc_dir["rank"] <= TOP_K].copy()

    return topk

cooc3 = build_item_cooc(b_train, min_cooc=3)
cooc2 = build_item_cooc(b_train, min_cooc=2)

# fallback: usar cooc3 quando existe; caso contrário usar cooc2 (para produtos sem vizinhos no cooc3)
products3 = set(cooc3["product_id"].unique())
missing_in_3 = set(cooc2["product_id"].unique()) - products3

cooc_final = pd.concat(
    [cooc3, cooc2[cooc2["product_id"].isin(missing_in_3)]],
    ignore_index=True
)

# re-rank final por product_id
cooc_final = cooc_final.sort_values(["product_id","cosine","cooc_count"], ascending=[True, False, False])
cooc_final["rank"] = cooc_final.groupby("product_id").cumcount() + 1
cooc_final = cooc_final[cooc_final["rank"] <= TOP_K].copy()

print("[B] item_cooc_final:", cooc_final.shape)
print("[B] unique products w/ neighbors:", cooc_final["product_id"].nunique())
print("[B] coverage product_id in train:", cooc_final["product_id"].isin(train_items).mean())
print("[B] coverage neighbor_id in train:", cooc_final["neighbor_id"].isin(train_items).mean())

cooc_final.to_csv(OUT_DIR / "ITEM_COOC_TOPK_FALLBACK_TRAIN.csv", index=False, encoding="utf-8")
```

## Content Candidates

```python
TOP_K_CONTENT = 100  

# --- stats no treino para ranking dos vizinhos ---
ti = train_interactions.copy()
ti["product_id"] = ti["product_id"].astype(int)
ti["customer_key"] = ti["customer_key"].astype(str)
ti["strength"] = pd.to_numeric(ti["strength"], errors="coerce").fillna(0.0)

item_stats = (
    ti.groupby("product_id")
      .agg(
          neighbor_popularity=("customer_key", "nunique"),
          neighbor_total_revenue=("strength", "sum")
      )
      .reset_index()
)

# --- metadata dos items (só treino) ---
items_meta = df_items_train.copy()
items_meta["product_id"] = items_meta["product_id"].astype(int)

for col in ["brand", "hs_code"]:
    if col not in items_meta.columns:
        raise ValueError(f"df_items_train precisa da coluna '{col}' para content neighbors.")

# juntar stats ao item master
items_meta = items_meta.merge(item_stats, on="product_id", how="left")
items_meta["neighbor_popularity"] = items_meta["neighbor_popularity"].fillna(0).astype(int)
items_meta["neighbor_total_revenue"] = items_meta["neighbor_total_revenue"].fillna(0.0)

# lookup rápido do neighbor (evita confusões de colunas)
neighbor_lookup = (
    items_meta[["product_id", "brand", "hs_code", "neighbor_popularity", "neighbor_total_revenue"]]
    .rename(columns={
        "product_id": "neighbor_id",
        "brand": "neighbor_brand",
        "hs_code": "neighbor_hs_code"
    })
)

def build_neighbors_by(feature_col: str, reason_label: str):
    base = items_meta[["product_id", feature_col]].dropna().copy()

    # self-join: produtos que partilham o mesmo feature
    pairs = base.merge(base, on=feature_col, suffixes=("_src", "_nbr"))
    pairs = pairs[pairs["product_id_src"] != pairs["product_id_nbr"]].copy()

    # renomear IDs de forma explícita
    pairs = pairs.rename(columns={
        "product_id_src": "product_id",
        "product_id_nbr": "neighbor_id"
    })

    # juntar metadata + stats do neighbor
    pairs = pairs.merge(neighbor_lookup, on="neighbor_id", how="left")

    pairs["reason"] = reason_label

    # ordenar por qualidade do neighbor (pop + revenue)
    pairs = pairs.sort_values(
        ["product_id", "neighbor_popularity", "neighbor_total_revenue"],
        ascending=[True, False, False]
    )

    # rank e topK por product_id (dentro do reason)
    pairs["rank"] = pairs.groupby("product_id").cumcount() + 1
    pairs = pairs[pairs["rank"] <= TOP_K_CONTENT].copy()

    return pairs[[
        "product_id", "neighbor_id", "reason", "rank",
        "neighbor_brand", "neighbor_hs_code",
        "neighbor_total_revenue", "neighbor_popularity"
    ]]

# construir para hs_code e brand
content_hs = build_neighbors_by("hs_code", "hs_code")
content_br = build_neighbors_by("brand", "brand")

item_content_train = pd.concat([content_hs, content_br], ignore_index=True)

# QC
train_items = set(train_interactions["product_id"].astype(int).unique())
print("[C] item_content_train shape:", item_content_train.shape)
print("[C] unique products with >=1 neighbor:", item_content_train["product_id"].nunique())
print("[C] avg neighbors per product:", item_content_train.groupby("product_id").size().mean())
print("[C] reason split:\n", item_content_train["reason"].value_counts(normalize=True).round(3))
print("[C] coverage product_id in train:", item_content_train["product_id"].isin(train_items).mean())
print("[C] coverage neighbor_id in train:", item_content_train["neighbor_id"].isin(train_items).mean())

print("Head(10):")
print(item_content_train.head(10))

item_content_train.to_csv(OUT_DIR / "ITEM_CONTENT_NEIGHBORS_TRAIN.csv", index=False, encoding="utf-8")
print("[SAVED] ITEM_CONTENT_NEIGHBORS_TRAIN.csv")
```
