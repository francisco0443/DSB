```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
ENCODINGS_TO_TRY = ["utf-8", "utf-8-sig", "latin1", "cp1252"]
DELIMITERS_TO_TRY = [",", ";", "\t", "|"]

def read_csv_safely(filename, nrows=None):
    """
    Lê um CSV tentando vários encodings e delimitadores.
    filename: nome do ficheiro (string) ou Path relativo a DATA_DIR
    nrows: se quiseres ler só X linhas (por ex. para testes)
    """
    path = DATA_DIR / filename
    last_error = None
    
    for enc in ENCODINGS_TO_TRY:
        for delim in DELIMITERS_TO_TRY:
            try:
                print(f"[INFO] A ler {path.name} com encoding='{enc}', delimiter='{repr(delim)}'")
                df = pd.read_csv(path, encoding=enc, delimiter=delim, nrows=nrows)
                print(f"[SUCCESS] Carregado com sucesso!")
                return df
            except (UnicodeDecodeError, pd.errors.ParserError) as e:
                last_error = e
                continue
    
    # último recurso: ignora caracteres marados e usa comma
    print(f"[AVISO] A ler {path.name} com utf-8, errors='ignore', delimiter=','")
    df = pd.read_csv(path, encoding="utf-8", errors="ignore", delimiter=",", nrows=nrows)
    return df
```

## Master Table

```python
# Análise RFM (usando created_at de order_item para recency)

order_col = "entity_id"

# 0) Recalcular total por encomenda a partir de order_item
#    -> excluir itens "free" (ex: gift-*, influencer-card, transport*, replacement*, etc.)
#    -> usar uma métrica de receita por linha (line_revenue) em vez de base_price

oi = df_order_item.copy()

# 0.1) garantir que created_at é datetime
oi["created_at"] = pd.to_datetime(oi["created_at"], errors="coerce")

# 0.2) garantir numéricos 
for c in ["base_row_total_incl_tax", "base_row_total", "base_price", "qty_ordered", "base_discount_amount"]:
    if c in oi.columns:
        oi[c] = pd.to_numeric(oi[c], errors="coerce").fillna(0.0)

# 0.3) escolher métrica mais robusta para receita por linha
if "base_row_total_incl_tax" in oi.columns:
    oi["line_revenue"] = oi["base_row_total_incl_tax"]
elif "base_row_total" in oi.columns:
    oi["line_revenue"] = oi["base_row_total"]
else:
    # fallback: base_price * qty (melhor do que apenas base_price)
    oi["line_revenue"] = oi["base_price"] * oi["qty_ordered"]

# 0.4) manter apenas itens pagos para definição de compra real
oi_paid = oi[oi["line_revenue"] > 0].copy()

# 0.5) total por encomenda (só itens pagos)
order_amounts = (
    oi_paid.groupby("order_id")["line_revenue"]
    .sum()
    .rename("order_amount_items")
)

# 1) última data por encomenda (só itens pagos)
order_last = (
    oi_paid.groupby("order_id")["created_at"]
    .max()
    .rename("order_last_created_at")
)

# 2) juntar data e total por encomenda à tabela de encomendas (entity_id -> order_id)
orders_with_item_dates = orders_valid.merge(order_last, left_on="entity_id", right_on="order_id", how="inner")
orders_with_item_dates = orders_with_item_dates.merge(order_amounts, left_on="entity_id", right_on="order_id", how="inner")

# 2.0) manter apenas encomendas com valor positivo (compra real)
orders_with_item_dates = orders_with_item_dates[orders_with_item_dates["order_amount_items"] > 0].copy()

# trazer o email do sales_flat_order_address via shipping_address_id -> entity_id
orders_with_item_dates = orders_with_item_dates.merge(
    df_order_addr[["entity_id","email"]],
    left_on="shipping_address_id",
    right_on="entity_id",
    how="left",
    suffixes=("","_addr")
)

# 2.1) Guarda orders de clientes sem email e sem customer_id noutra tabela para análise futura
orders_no_id = orders_with_item_dates[
    orders_with_item_dates["email"].isna() & orders_with_item_dates["customer_id"].isna()
].copy()
print("Orders without email and customer_id:", orders_no_id.shape)

# 2.2) Remover orders de clientes sem email e sem customer_id
orders_with_item_dates = orders_with_item_dates[
    ~(orders_with_item_dates["email"].isna() & orders_with_item_dates["customer_id"].isna())
].copy()
print("Orders after removing those without email and customer_id:", orders_with_item_dates.shape)

# 2.3) criar identificador de cliente com fallback row-by-row: email se existir, senão customer_id
# (evita o caso customer_key == 'nan' em string)
orders_with_item_dates["email"] = orders_with_item_dates["email"].replace("nan", np.nan)
orders_with_item_dates["customer_key"] = orders_with_item_dates["email"].fillna(orders_with_item_dates["customer_id"]).astype(str)
orders_with_item_dates = orders_with_item_dates[orders_with_item_dates["customer_key"] != "nan"].copy()

customer_col = "customer_key"  # usar o identificador unificado para agrupar

# 3) snapshot_date: dia após a última data de criação de item pago
snapshot_date = orders_with_item_dates["order_last_created_at"].max() + pd.Timedelta(days=1)

# 4) agregar por cliente usando timestamp do item pago para recency e usar order_amount_items para monetary
rfm = orders_with_item_dates.groupby(customer_col).agg({
    "order_last_created_at": lambda x: (snapshot_date - x.max()).days,  # recency em dias
    order_col: "nunique",                                                # frequency: encomendas distintas
    "order_amount_items": "sum"                                          # monetary: soma de revenue por ordem
})

# nomes padrões das colunas RFM
rfm.columns = ["recency", "frequency", "monetary"]

# pontuação simples R/F/M por quartis (1-4 cada)
try:
    rfm["r_score"] = pd.qcut(rfm["recency"], 4, labels=[4,3,2,1])
except Exception:
    rfm["r_score"] = pd.cut(rfm["recency"], bins=4, labels=[4,3,2,1])

try:
    rfm["f_score"] = pd.qcut(rfm["frequency"].rank(method="first"), 4, labels=[1,2,3,4])
except Exception:
    rfm["f_score"] = pd.cut(rfm["frequency"].rank(method="first"), bins=4, labels=[1,2,3,4])

try:
    rfm["m_score"] = pd.qcut(rfm["monetary"], 4, labels=[1,2,3,4])
except Exception:
    rfm["m_score"] = pd.cut(rfm["monetary"], bins=4, labels=[1,2,3,4])

# substituir NaN por 1 antes de converter para int (1 = quartil mais baixo)
rfm[["r_score","f_score","m_score"]] = rfm[["r_score","f_score","m_score"]].fillna(1).astype(int)

# soma das três pontuações RFM
rfm["rfm_score"] = rfm[["r_score","f_score","m_score"]].sum(axis=1)

# dataframe com rfm 
rfm["recency"] = rfm["recency"].fillna(0).astype(int)
rfm["frequency"] = rfm["frequency"].fillna(0).astype(int)
rfm["monetary"] = rfm["monetary"].fillna(0).astype(float)

df_rfm = rfm.reset_index()

print("RFM table built:", df_rfm.shape)
print(df_rfm.head(8))
print("\nRFM summary:")
print(df_rfm[["recency","frequency","monetary","rfm_score"]].describe())

# sanity check: quantos clientes ainda têm monetary == 0?
print("\nCustomers with monetary == 0:", (df_rfm["monetary"] == 0).sum())
```

## Customer Data Frame

```python
# Criar df customer level
# Cópia da tabela RFM sem os r_score, f_score, m_score
df_customer = df_rfm.drop(columns=["r_score","f_score","m_score"]).copy()
print("Customer-level table created:", df_customer.shape)
print(df_customer.head(8))
```

```python
# Adicionar discount_sensitivity
# Calculado através de (Soma de descontos aplicados / Soma de preços base) por customer
# Primeiro, calcular descontos e preços base por order_item
oi_paid["base_discount_amount"] = pd.to_numeric(oi_paid["base_discount_amount"], errors="coerce").fillna(0.0)
oi_paid["base_price"] = pd.to_numeric(oi_paid["base_price"], errors="coerce").fillna(0.0)   
order_item_discounts = oi_paid.groupby("order_id").agg({
    "base_discount_amount": "sum",
    "base_price": "sum"
}).rename(columns={
    "base_discount_amount": "total_discount",
    "base_price": "total_base_price"
})
# Agora, juntar à tabela de encomendas para ter customer_key
orders_with_discounts = orders_with_item_dates.merge(
    order_item_discounts,
    left_on="entity_id",
    right_on="order_id",
    how="left"
)
# Agora, agregar por customer_key para calcular discount_sensitivity
customer_discounts = orders_with_discounts.groupby("customer_key").agg({
    "total_discount": "sum",
    "total_base_price": "sum"
}).reset_index()
# Calcular discount_sensitivity
customer_discounts["discount_sensitivity"] = np.where(
    customer_discounts["total_base_price"] > 0,
    customer_discounts["total_discount"] / customer_discounts["total_base_price"],
    0.0
)
# Juntar à tabela customer-level
df_customer = df_customer.merge(
    customer_discounts[["customer_key", "discount_sensitivity"]],
    on="customer_key",
    how="left"
)
print("Customer-level table after adding discount_sensitivity:", df_customer.shape)
print(df_customer.head(8))
# Verificar missing values na nova coluna
print("Missing values in discount_sensitivity:", df_customer["discount_sensitivity"].isna().sum())
```

```python
# Adicionar coluna return_rate
# Calculado através de (Nº de orders com refund / Nº total de orders) por customer
# Primeiro, criar coluna has_refund em orders_with_item_dates
orders_with_item_dates["has_refund"] = orders_with_item_dates["total_refunded"] > 0
# Agora, agregar por customer_key para calcular return_rate
customer_returns = orders_with_item_dates.groupby("customer_key").agg({
    "has_refund": "sum",
    "entity_id": "nunique"
}).rename(columns={
    "has_refund": "num_orders_with_refund",
    "entity_id": "total_num_orders"
}).reset_index()
# Calcular return_rate
customer_returns["return_rate"] = np.where(
    customer_returns["total_num_orders"] > 0,
    customer_returns["num_orders_with_refund"] / customer_returns["total_num_orders"],
    0.0
)
# Juntar à tabela customer-level
df_customer = df_customer.merge(
    customer_returns[["customer_key", "return_rate"]],
    on="customer_key",
    how="left"
)
print("Customer-level table after adding return_rate:", df_customer.shape)
print(df_customer.head(8))
# Verificar missing values na nova coluna
print("Missing values in return_rate:", df_customer["return_rate"].isna().sum())
print(df_customer["return_rate"].describe())
```

```python
import re

prod_hs_opt = df_prod_int[df_prod_int["attribute_id"] == 168][["entity_id", "value"]].copy()

# IDs numéricos (evitar float)
prod_hs_opt["entity_id"] = pd.to_numeric(prod_hs_opt["entity_id"], errors="coerce").astype("Int64")
prod_hs_opt["value"]     = pd.to_numeric(prod_hs_opt["value"], errors="coerce").astype("Int64")  # option_id

# 2) map option_id -> text (eav_option) escolhendo 1 linha por option_id
#    (preferir store_id==0; senão menor store_id)
eav = df_eav_option[["option_id", "store_id", "value"]].copy()
eav["option_id"] = pd.to_numeric(eav["option_id"], errors="coerce").astype("Int64")
eav["store_id"]  = pd.to_numeric(eav["store_id"], errors="coerce").astype("Int64")

# 3) join para obter HS "texto/categoria"
prod_hs = prod_hs_opt.merge(eav, left_on="value", right_on="option_id", how="left")

# HS label (categoria textual) — o mais útil para clustering/recommender
prod_hs["hs_label"] = prod_hs["value_y"]

# (opcional) extrair dígitos HS se existirem no texto
def extract_hs_digits(text):
    if pd.isna(text):
        return np.nan
    m = re.search(r"(\d{8,10})", str(text))
    return m.group(1) if m else np.nan

prod_hs["hs_code"] = prod_hs["hs_label"].apply(extract_hs_digits)

# mapping final product_id -> hs_label (+ hs_code opcional)
prod_hs_map = prod_hs[["entity_id", "hs_label", "hs_code"]].drop_duplicates(subset=["entity_id"]).copy()

# 4) attach HS ao catálogo (prod_final)
prod_final["entity_id"] = pd.to_numeric(prod_final["entity_id"], errors="coerce").astype("Int64")
prod_final = prod_final.merge(prod_hs_map, on="entity_id", how="left")

# 5) attach HS aos itens pagos (oi_paid)
oi_paid["product_id"] = pd.to_numeric(oi_paid["product_id"], errors="coerce").astype("Int64")

# Antes de juntar o HS ao oi_paid: garantir que não existem colunas HS antigas
oi_paid = oi_paid.drop(columns=[c for c in ["hs_code", "hs_label"] if c in oi_paid.columns], errors="ignore")

oi_paid = oi_paid.merge(
    prod_hs_map.rename(columns={"entity_id": "product_id"}),
    on="product_id",
    how="left"
)

# 6) favorite_hs por customer
items_with_customer = oi_paid.merge(
    orders_with_item_dates[["entity_id", "customer_key"]],
    left_on="order_id",
    right_on="entity_id",
    how="left"
)

items_with_customer = items_with_customer.dropna(subset=["hs_code"])

hs_counts = (
    items_with_customer.groupby(["customer_key", "hs_code"])
    .size()
    .rename("cnt")
    .reset_index()
    .sort_values(["customer_key", "cnt"], ascending=[True, False])
)

favorite_hs = hs_counts.drop_duplicates(subset=["customer_key"], keep="first").copy()
favorite_hs = favorite_hs.rename(columns={"hs_code": "favorite_hs", "cnt": "favorite_hs_count"})

df_customer = df_customer.merge(favorite_hs, on="customer_key", how="left")

print("Customer-level table after adding favorite_hs:", df_customer.shape)
print(df_customer.head(8))
# Verificar missing values na nova coluna
print("Missing values in favorite_hs:", df_customer["favorite_hs"].isna().sum())
```

```python
# Tenure: dias desde a primeira compra (order_last_created_at mínimo) até ao snapshot_date
snapshot_date = orders_with_item_dates["order_last_created_at"].max() + pd.Timedelta(days=1)

customer_first = (
    orders_with_item_dates.groupby("customer_key")["order_last_created_at"]
    .min()
    .rename("first_purchase_date")
    .reset_index()
)

customer_first["tenure_days"] = (snapshot_date - customer_first["first_purchase_date"]).dt.days

df_customer = df_customer.merge(
    customer_first[["customer_key", "tenure_days"]],
    on="customer_key",
    how="left"
)

print("Customer-level table after adding tenure_days:", df_customer.shape)
print(df_customer.head(8))
print("Missing values in tenure_days:", df_customer["tenure_days"].isna().sum())
```

```python
# Regularidade: média de dias entre compras por cliente (paid orders)
tmp = orders_with_item_dates[["customer_key", "order_last_created_at"]].dropna().copy()
tmp = tmp.sort_values(["customer_key", "order_last_created_at"])

tmp["prev_date"] = tmp.groupby("customer_key")["order_last_created_at"].shift(1)
tmp["interpurchase_days"] = (tmp["order_last_created_at"] - tmp["prev_date"]).dt.days

avg_interpurchase = (
    tmp.groupby("customer_key")["interpurchase_days"]
    .mean()
    .rename("avg_interpurchase_days")
    .reset_index()
)

df_customer = df_customer.merge(avg_interpurchase, on="customer_key", how="left")
df_customer["avg_interpurchase_days"] = df_customer["avg_interpurchase_days"].fillna(0).astype(float)

print("Customer-level table after adding avg_interpurchase_days:", df_customer.shape)
print(df_customer.head(8))
print("Missing values in avg_interpurchase_days:", df_customer["avg_interpurchase_days"].isna().sum())
```

```python
#Ver HS único por cliente
# Garantir que hs_code existe no oi_paid
if "hs_code" not in oi_paid.columns:
    raise ValueError("oi_paid não tem hs_code. Garante que fizeste o merge com prod_hs_map e normalizaste hs_code.")

# ligar itens pagos ao customer_key via order_id -> orders_with_item_dates.entity_id
items_with_customer = oi_paid.merge(
    orders_with_item_dates[["entity_id", "customer_key"]],
    left_on="order_id",
    right_on="entity_id",
    how="left"
)

# unique_hs: nº de HS distintos por cliente
hs_div = (
    items_with_customer.dropna(subset=["hs_code"])
    .groupby("customer_key")["hs_code"]
    .nunique()
    .rename("unique_hs")
    .reset_index()
)

df_customer = df_customer.merge(hs_div, on="customer_key", how="left")
df_customer["unique_hs"] = df_customer["unique_hs"].fillna(0).astype(int)

# favorite_hs_share: favorite_hs_count / total itens pagos (linhas) do cliente
total_paid_items = (
    items_with_customer.groupby("customer_key")
    .size()
    .rename("total_paid_item_lines")
    .reset_index()
)

df_customer = df_customer.merge(total_paid_items, on="customer_key", how="left")

# precisa de favorite_hs_count existir (já tens)
df_customer["favorite_hs_share"] = np.where(
    df_customer["total_paid_item_lines"] > 0,
    df_customer["favorite_hs_count"] / df_customer["total_paid_item_lines"],
    0.0
)

print("Customer-level table after adding unique_hs + favorite_hs_share:", df_customer.shape)
print(df_customer.head(8))
print("Missing values in unique_hs:", df_customer["unique_hs"].isna().sum())
print("Missing values in favorite_hs_share:", df_customer["favorite_hs_share"].isna().sum())
```

```python
# avg_items_per_order: nº médio de linhas de item pagas por encomenda do cliente
items_per_order = (
    oi_paid.groupby("order_id")
    .size()
    .rename("paid_item_lines")
    .reset_index()
)

orders_items = orders_with_item_dates.merge(
    items_per_order,
    left_on="entity_id",
    right_on="order_id",
    how="left"
)

cust_items = (
    orders_items.groupby("customer_key")["paid_item_lines"]
    .mean()
    .rename("avg_items_per_order")
    .reset_index()
)

df_customer = df_customer.merge(cust_items, on="customer_key", how="left")
df_customer["avg_items_per_order"] = df_customer["avg_items_per_order"].fillna(0).astype(float)

print("Customer-level table after adding avg_items_per_order:", df_customer.shape)
print(df_customer.head(8))
print("Missing values in avg_items_per_order:", df_customer["avg_items_per_order"].isna().sum())
```

```python
# std_order_value: desvio padrão do valor das encomendas por cliente (paid orders)
cust_order_std = (
    orders_with_item_dates.groupby("customer_key")["order_amount_items"]
    .std()
    .rename("std_order_value")
    .reset_index()
)

df_customer = df_customer.merge(cust_order_std, on="customer_key", how="left")

# clientes com 1 order => std = NaN, faz sentido preencher 0
df_customer["std_order_value"] = df_customer["std_order_value"].fillna(0).astype(float)

print("Customer-level table after adding std_order_value:", df_customer.shape)
print(df_customer.head(8))
print("Missing values in std_order_value:", df_customer["std_order_value"].isna().sum())
```

```python
# alinhar tipos e trazer region + country_id da tabela de endereços
orders_with_item_dates["shipping_address_id"] = orders_with_item_dates["shipping_address_id"].astype("Int64")
df_order_addr["entity_id"] = df_order_addr["entity_id"].astype("Int64")

# criar séries mapeáveis (index = entity_id)
addr_country = df_order_addr.set_index("entity_id")["country_id"]

# mapear para a orders_with_item_dates
orders_with_item_dates["country_id"] = orders_with_item_dates["shipping_address_id"].map(addr_country)


# país da encomenda mais recente por cliente
country_last = (
    orders_with_item_dates.sort_values(["customer_key", "order_last_created_at"])
    .groupby("customer_key")["country_id"]
    .last()
    .rename("country_last")
)

# país mais frequente por cliente (moda)
country_mode = (
    orders_with_item_dates.groupby("customer_key")["country_id"]
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
    .rename("country_mode")
)

# juntar as features de país à tabela final de clientes
df_customer = df_customer.merge(country_last.reset_index(), on="customer_key", how="left")
df_customer = df_customer.merge(country_mode.reset_index(), on="customer_key", how="left")

print(df_customer.head(8))
```

```python
def pct(x):
    return round(float(x) * 100, 2)

def safe_dt(s):
    return pd.to_datetime(s, errors="coerce", utc=False)

DATA_DIR = Path(r"CSVTables")

df_customer_personas = read_csv_safely("DF_CUSTOMER_PERSONAS.csv")


print("\n[1] df_customer_personas shape:", df_customer_personas.shape)
print("[1] df_customer_personas columns:\n", list(df_customer_personas.columns))


print("\n[2] interactions shape:", interactions.shape)
print("[2] Missing values in key columns:")
print(interactions[["customer_key", "product_id"]].isna().sum())

# Coverage: % customers + % items (interactions vs users/items)

users_set = set(df_customer_personas["customer_key"].astype(str))
int_users = interactions["customer_key"].astype(str)

items_set = set(pd.to_numeric(df_items["product_id"], errors="coerce").dropna().astype("Int64").astype(int))
int_items = pd.to_numeric(interactions["product_id"], errors="coerce").dropna().astype("Int64").astype(int)

n_int_users = int_users.nunique()
n_known_users = int(int_users[int_users.isin(users_set)].nunique())

n_int_items = int_items.nunique()
n_known_items = int(pd.Series(int_items)[pd.Series(int_items).isin(items_set)].nunique())

print("\n[3] Coverage (interactions -> known tables)")
print(f"[3] Customers in interactions: {n_int_users} | matched in df_customer_personas: {n_known_users} ({pct(n_known_users/max(n_int_users,1))}%)")
print(f"[3] Items in interactions:     {n_int_items} | matched in df_items:            {n_known_items} ({pct(n_known_items/max(n_int_items,1))}%)")

# Strength distribution
print("\n[4] interactions['strength'] describe:")
if "strength" not in interactions.columns:
    raise ValueError("interactions não tem coluna 'strength'. Confirma o nome (ex: 'line_revenue' ou 'interaction_strength').")
print(interactions["strength"].describe(percentiles=[.5, .9, .95, .99]))

# df_items missing rates em brand/hs_code
print("\n[5] Missing rate em df_items (brand, hs_code):")
cols_check = [c for c in ["brand", "hs_code"] if c in df_items.columns]
if not cols_check:
    print("[5] df_items não tem 'brand' nem 'hs_code'. Confirma colunas disponíveis:", list(df_items.columns))
else:
    miss_rates = df_items[cols_check].isna().mean().sort_values(ascending=False)
    for c, r in miss_rates.items():
        print(f"[5] {c}: {pct(r)}% missing")

# Distribuição de clusters (RB4) no df_customer_personas
print("\n[6] cluster distribution (RB4):")
cl_col = "cluster_rb4" if "cluster_rb4" in df_customer_personas.columns else None
if cl_col is None:
    print("[6] Não encontrei 'cluster_rb4' em df_customer_personas. Colunas:", list(df_customer_personas.columns))
else:
    print(df_customer_personas[cl_col].value_counts(dropna=False, normalize=True).round(4))

# Nº médio de items distintos por user (interactions)
print("\n[7] Unique items per user (from interactions):")
tmp = interactions.copy()
tmp["customer_key"] = tmp["customer_key"].astype(str)
tmp["product_id"] = pd.to_numeric(tmp["product_id"], errors="coerce")
tmp = tmp.dropna(subset=["customer_key", "product_id"])
items_per_user = tmp.groupby("customer_key")["product_id"].nunique()
print(items_per_user.describe(percentiles=[.5, .9, .95, .99]))

# Min/max de last_ts (para split temporal)
print("\n[8] last_ts min/max:")
ts_col = "last_ts" if "last_ts" in interactions.columns else None
if ts_col is None:
    print("[8] Não encontrei 'last_ts' em interactions. Colunas:", list(interactions.columns))
else:
    ts = safe_dt(interactions[ts_col])
    print("[8] last_ts invalid (NaT):", int(ts.isna().sum()))
    print("[8] last_ts min:", ts.min())
    print("[8] last_ts max:", ts.max())
```
