# Data Reading

```python
# Ranking plugável — LightGBM (LambdaMART) skeleton


import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split
import lightgbm as lgb
import joblib
import time

DATA_DIR = Path(r"CSVTables\Recommender\TreinoTeste")

train_interactions = pd.read_csv(DATA_DIR / "train_interactions.csv")
test_gt = pd.read_csv(DATA_DIR / "test_gt.csv")
```

```python
DATA_DIR = Path(r"CSVTables\Recommender") 

df_items = pd.read_csv(DATA_DIR / "DF_ITEMS.csv", delimiter=';')
df_customers = pd.read_csv(DATA_DIR / "DF_CUSTOMER_PERSONAS.csv", delimiter=';')
```

```python
DATA_DIR = Path(r"CSVTables\Recommender\CG") 

persona_top = pd.read_csv(DATA_DIR / "PERSONA_TOP_ITEMS.csv", delimiter=';')
item_cooc = pd.read_csv(DATA_DIR / "ITEM_COOC_TOPK_FALLBACK.csv", delimiter=';')
item_content = pd.read_csv(DATA_DIR / "ITEM_CONTENT_NEIGHBORS.csv", delimiter=';')
```

```python
# standardize column names (lowercase)
train_interactions.columns = [c.lower() for c in train_interactions.columns]
test_gt.columns = [c.lower() for c in test_gt.columns]
df_customers.columns = [c.lower() for c in df_customers.columns]
df_items.columns = [c.lower() for c in df_items.columns]

# ensure strength_log exists (if not, compute)
if "strength_log" not in train_interactions.columns:
    train_interactions["strength_log"] = np.log1p(train_interactions["strength"].fillna(0.0))


print(persona_top.head())
print(item_cooc.head())
print(item_content.head())
```

# Recommender

```python

# ---------------------------
# 1) Build candidate set per user
# 
# Here we construct candidates by:
#  - persona_top (persona -> top items)
#  - for each user, neighbours from item_cooc and item_content for items in their train history
#  - add global fallback top-M items if needed
# ---------------------------

# quick persona map
persona_map = df_customers.drop_duplicates(subset=["customer_key"]).set_index("customer_key")["persona"].to_dict()

# precompute global ranked by strength_log
pop_global = train_interactions.groupby("product_id")["strength_log"].sum().sort_values(ascending=False)
global_ranked = list(pop_global.index)
GLOBAL_FALLBACK_M = 200

# user train history (unique items)
user_items = train_interactions.groupby("customer_key")["product_id"].apply(list).to_dict()

# helper: persona_top dict
persona_top_dict = (
    persona_top.sort_values(["persona", "rank"])
    .groupby("persona")["product_id"]
    .apply(list)
    .to_dict()
)

# cooc neighbors mapping: for each product -> list of (neighbor, score)
cooc_map = {}
if "neighbor_id" in item_cooc.columns:
    grp = item_cooc.sort_values(["product_id", "cosine", "cooc_count"], ascending=[True, False, False])
    cooc_map = grp.groupby("product_id").apply(lambda df: df[["neighbor_id","cosine"]].values.tolist()).to_dict()

# content neighbors mapping
content_map = {}
if "neighbor_id" in item_content.columns:
    grp = item_content.sort_values(["product_id","rank"])
    content_map = grp.groupby("product_id").apply(lambda df: df[["neighbor_id","rank"]].values.tolist()).to_dict()

# Build candidates list (this uses a loop over users-acceptable for training data prep)
cands = []
USERS = list(test_gt["customer_key"])  # only build candidates for users we will evaluate/train on

for u in USERS:
    # persona candidates
    persona = persona_map.get(u, None)
    p_items = persona_top_dict.get(persona, [])[:50] if persona is not None else []
    for pid in p_items:
        cands.append((u, int(pid), "persona", 1.0))  # score 1.0 is placeholder (normalized later)
    # cooc from items in user history
    his = user_items.get(u, [])
    for it in his:
        neighs = cooc_map.get(it, [])
        for n, score in neighs[:50]:
            cands.append((u, int(n), "cooc", float(score)))
    # content neighbors
    for it in his:
        neighs = content_map.get(it, [])
        for n, rank in neighs[:50]:
            # convert rank -> score (higher for smaller rank)
            score = 1.0 / (1 + float(rank))
            cands.append((u, int(n), "content", score))
    # fallback global top items
    for pid in global_ranked[:GLOBAL_FALLBACK_M]:
        cands.append((u, int(pid), "global", float(pop_global.loc[pid])))
# consolidate to DataFrame
candidates_raw = pd.DataFrame(cands, columns=["customer_key","product_id","source","raw_score"])
print(candidates_raw)
```

```python

# ---------------------------
# 2) Aggregate and create features per (user, product)
# ---------------------------
# normalize raw_score per (customer, source) using min-max to balance sources per user
def min_max_group(g):
    mn = g.min()
    mx = g.max()
    if mx == mn:
        return (g - mn) * 0.0
    return (g - mn) / (mx - mn)

candidates_raw["raw_score"] = candidates_raw["raw_score"].fillna(0.0)
candidates_raw["raw_score_norm"] = candidates_raw.groupby(["customer_key","source"])["raw_score"].transform(min_max_group)

print(candidates_raw)
# aggregate: for each (user, product) compute aggregated features
agg = candidates_raw.pivot_table(
    index=["customer_key","product_id"],
    columns="source",
    values="raw_score_norm",
    aggfunc="max",
    fill_value=0
).reset_index()
# fill missing source columns
for s in ["persona","cooc","content","global"]:
    if s not in agg.columns:
        agg[s] = 0.0

# add features: number of sources that suggested the item, sum of scores
agg["n_sources"] = (agg[["persona","cooc","content","global"]] > 0).sum(axis=1)
agg["sum_source_scores"] = agg[["persona","cooc","content","global"]].sum(axis=1)

# ---------------------------
# 3) Join item features and simple user features (RFM-like)
# ---------------------------
# item features
df_items_small = df_items.rename(columns={c:c.lower() for c in df_items.columns})
df_items_small = df_items_small[["product_id","brand","hs_code","total_revenue","price_median","popularity"]].copy()
df_items_small["product_id"] = df_items_small["product_id"].astype(int)

agg = agg.merge(df_items_small, on="product_id", how="left")

# recomputar user features de forma robusta
train_inter = train_interactions.copy()
if "last_ts" in train_inter.columns:
    train_inter["last_ts"] = pd.to_datetime(train_inter["last_ts"], errors="coerce")
    last_ts_user = train_inter.groupby("customer_key")["last_ts"].max().rename("last_ts")
    freq_user = train_inter.groupby("customer_key")["product_id"].nunique().rename("n_items")
    mon_user = train_inter.groupby("customer_key")["strength_log"].sum().rename("mon")
    user_feats = pd.concat([last_ts_user, freq_user, mon_user], axis=1).reset_index()
    maxdate = train_inter["last_ts"].max()
    user_feats["recency_days"] = (maxdate - user_feats["last_ts"]).dt.days.fillna(9999)
    user_feats = user_feats[["customer_key","n_items","mon","recency_days"]]
else:
    # cria um df vazio com colunas esperadas (será imputado)
    user_feats = pd.DataFrame(columns=["customer_key","n_items","mon","recency_days"])

# merge e imputação (valores defensivos)
agg = agg.merge(user_feats, on="customer_key", how="left")
agg["n_items"] = agg["n_items"].fillna(0).astype(int)
agg["mon"] = agg["mon"].fillna(0.0)
agg["recency_days"] = agg["recency_days"].fillna(9999).astype(int)

# ---------------------------
# 4) Label creation using test_gt (1 if product in gt)
# ---------------------------
gt_map = test_gt.set_index("customer_key")["gt_items"].to_dict()
def label_row(row):
    u = row["customer_key"]
    p = row["product_id"]
    rel = set(gt_map.get(u, []))
    return 1 if p in rel else 0

agg["label"] = agg.apply(label_row, axis=1)

# optional: drop candidates where product not in df_items (na features) - but better keep and impute
agg["brand"] = agg["brand"].fillna("unknown")
agg["hs_code"] = agg["hs_code"].fillna("unknown")
agg["total_revenue"] = agg["total_revenue"].fillna(0.0)
agg["price_median"] = agg["price_median"].fillna(0.0)
agg["popularity"] = agg["popularity"].fillna(0.0)


```

```python
print(agg.head())
print(agg["customer_key"].nunique())
print(agg.isna().sum())
```

```python
import random
import numpy as np

RANDOM_STATE = 42

# ---------------------------
# reproducibility (minimal, as requested)
# ---------------------------
random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)

# ---------------------------
# 5) Train / Validation split by users (to avoid leakage)
# ---------------------------
unique_users = agg["customer_key"].unique()
train_users, val_users = train_test_split(unique_users, test_size=0.2, random_state=RANDOM_STATE)
train_df = agg[agg["customer_key"].isin(train_users)].reset_index(drop=True)
val_df = agg[agg["customer_key"].isin(val_users)].reset_index(drop=True)

FEATURE_COLS = ["persona","cooc","content","global","n_sources","sum_source_scores",
                "total_revenue","price_median","popularity","n_items","mon","recency_days"]

# ---- recommended: compute categorical codes on the full agg to ensure consistent mapping ----
# minimal change: factorize on whole dataset then split
agg["brand_code"] = pd.factorize(agg["brand"].fillna("unknown"))[0]
agg["hs_code_code"] = pd.factorize(agg["hs_code"].fillna("unknown"))[0]

# reflect codes in train/val (they are derived from agg so both consistent)
train_df = train_df.merge(agg[["customer_key","product_id","brand_code","hs_code_code"]], on=["customer_key","product_id"], how="left")
val_df = val_df.merge(agg[["customer_key","product_id","brand_code","hs_code_code"]], on=["customer_key","product_id"], how="left")

FEATURE_COLS += ["brand_code","hs_code_code"]

# Ensure ordering by customer_key so group sizes line up with contiguous blocks
train_df = train_df.sort_values("customer_key").reset_index(drop=True)
val_df = val_df.sort_values("customer_key").reset_index(drop=True)

# prepare groups (counts per user)
def make_lgb_dataset(df):
    # IMPORTANT: df must be ordered by customer_key so group sizes correspond to contiguous rows
    X = df[FEATURE_COLS]
    y = df["label"].astype(int)
    groups = df.groupby("customer_key").size().tolist()
    return X, y, groups

X_train, y_train, group_train = make_lgb_dataset(train_df)
X_val, y_val, group_val = make_lgb_dataset(val_df)

# LightGBM datasets require concatenation per group ordering (we already enforced ordering)
lgb_train = lgb.Dataset(X_train, label=y_train, group=group_train)
lgb_val = lgb.Dataset(X_val, label=y_val, group=group_val, reference=lgb_train)

# ---------------------------
# 6) Train LightGBM Lambdarank
# ---------------------------
K_EVAL = 10

params = {
    "objective": "lambdarank",
    "metric": "ndcg",
    "ndcg_eval_at": [K_EVAL],
    "learning_rate": 0.05,
    "num_leaves": 64,
    "min_data_in_leaf": 20,
    "feature_fraction": 0.8,
    "bagging_fraction": 0.8,
    "bagging_freq": 5,
    "verbose": -1,
    # reproducibility-related seeds
    "seed": RANDOM_STATE,
    "bagging_seed": RANDOM_STATE,
    "feature_fraction_seed": RANDOM_STATE,
    # set threads to 1 for deterministic behaviour (may slow training)
    "num_threads": 1
}

start = time.time()
gbm = lgb.train(
    params,
    lgb_train,
    num_boost_round=1000,
    valid_sets=[lgb_train, lgb_val],
    valid_names=["train","valid"],
    callbacks=[
        lgb.early_stopping(stopping_rounds=50),
        lgb.log_evaluation(period=50)
    ]
)
end = time.time()
print(f"Training time: {end-start:.1f}s")

# save model
model_path = WORK_DIR / "lgb_rank_model.txt"
gbm.save_model(str(model_path))
joblib.dump(gbm, WORK_DIR / "lgb_rank_model.pkl")

# ---------------------------
# 7) Scoring / Building Top-K per user and evaluate
# ---------------------------
def score_and_get_recs(df_candidates, model, K=10):
    X = df_candidates[FEATURE_COLS]
    scores = model.predict(X, num_iteration=model.best_iteration)
    df_candidates = df_candidates.copy()
    df_candidates["score"] = scores
    recs = df_candidates.sort_values(["customer_key","score"], ascending=[True, False]).groupby("customer_key").head(K)
    reco_lists = recs.groupby("customer_key")["product_id"].apply(list).to_dict()
    return reco_lists, df_candidates

val_recs, val_with_scores = score_and_get_recs(val_df, gbm, K=K_EVAL)

# --- auxiliary: ground-truth map for validation users
gt_map = test_gt.set_index("customer_key")["gt_items"].to_dict()
val_gt_map = {u: gt_map.get(u, []) for u in val_df["customer_key"].unique()}

# ---------------------------
# Evaluation metrics (existing + added)
# ---------------------------
def precision_at_k(recs, gt_map, k=10):
    vals=[]
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        if len(rec)==0:
            vals.append(0.0); continue
        vals.append(len([x for x in rec if x in gt]) / k)
    return np.mean(vals)

def recall_at_k(recs, gt_map, k=10):
    vals=[]
    for u, gt in gt_map.items():
        if len(gt)==0:
            vals.append(0.0); continue
        rec = recs.get(u, [])[:k]
        vals.append(len([x for x in rec if x in gt]) / len(gt))
    return np.mean(vals)

def hit_rate_at_k(recs, gt_map, k=10):
    vals=[]
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        vals.append(1.0 if any(x in gt for x in rec) else 0.0)
    return np.mean(vals)

# new: nDCG@K
def ndcg_at_k(recs, gt_map, k=10):
    import math
    vals=[]
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        # binary relevance
        gains = [1 if r in gt else 0 for r in rec]
        dcg = sum([g / math.log2(i+2) for i, g in enumerate(gains)])  # i from 0 -> position i+1
        # ideal DCG: if there are R relevant items, best is putting them at top
        ideal_gains = [1]*min(len(gt), k)
        idcg = sum([g / math.log2(i+2) for i, g in enumerate(ideal_gains)]) if len(ideal_gains)>0 else 0.0
        vals.append(dcg / idcg if idcg>0 else 0.0)
    return np.mean(vals)

# new: MAP@K
def map_at_k(recs, gt_map, k=10):
    vals=[]
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        if len(gt)==0:
            vals.append(0.0); continue
        num_hits = 0
        s = 0.0
        for i, r in enumerate(rec, start=1):
            if r in gt:
                num_hits += 1
                s += num_hits / i  # precision@i
        denom = min(len(gt), k)
        vals.append(s / denom if denom>0 else 0.0)
    return np.mean(vals)

# candidate_recall@C (do conjunto de candidates antes do ranking)
candidate_candsets = val_df.groupby("customer_key")["product_id"].apply(set).to_dict()
def candidate_recall(candidate_sets, gt_map):
    vals=[]
    for u, gt in gt_map.items():
        cand = candidate_sets.get(u, set())
        if len(gt)==0:
            vals.append(0.0); continue
        vals.append(int(any(x in cand for x in gt)))
    return np.mean(vals)

# compute and print metrics
print("Val Precision@K:", precision_at_k(val_recs, val_gt_map, k=K_EVAL))
print("Val Recall@K:", recall_at_k(val_recs, val_gt_map, k=K_EVAL))
print("Val Hit@K:", hit_rate_at_k(val_recs, val_gt_map, k=K_EVAL))
print("Val nDCG@K:", ndcg_at_k(val_recs, val_gt_map, k=K_EVAL))
print("Val MAP@K:", map_at_k(val_recs, val_gt_map, k=K_EVAL))
print("Candidate_recall@C (validation candidate pool):", candidate_recall(candidate_candsets, val_gt_map))

# save recommendations and scored candidates
pd.DataFrame([{"customer_key":u,"recs":val_recs.get(u,[])} for u in val_recs]).to_parquet(WORK_DIR/"reco_hybrid_lgb_val.parquet", index=False)
val_with_scores.to_parquet(WORK_DIR/"val_candidates_scored.parquet", index=False)

print("Done.")
```
