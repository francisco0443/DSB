```python
from pathlib import Path
import pandas as pd
import numpy as np
import lightgbm as lgb

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

DATA_DIR = Path(r"CSVTables\Recommender")
SPLIT_DIR = Path(r"CSVTables\Recommender\Split_TrainTest")
CG_DIR    = Path(r"CSVTables\Recommender\Candidates_TRAIN")  # ajusta se necessário
OUT_DIR   = Path(r"CSVTables\Recommender\LGBM_OUT")
OUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# split data
train_interactions = pd.read_csv(SPLIT_DIR / "train_interactions.csv")
test_gt = pd.read_csv(SPLIT_DIR / "test_gt.csv")

# base tables (minimamente: personas + catálogo)
df_cust  = pd.read_csv(DATA_DIR / "DF_CUSTOMER_PERSONAS.csv", delimiter=";")
df_items = pd.read_csv(DATA_DIR / "DF_ITEMS.csv", delimiter=",")

# candidate tables computed on TRAIN only
persona_top = pd.read_csv(CG_DIR / "PERSONA_TOP_ITEMS_TRAIN.csv", delimiter=",")
item_cooc   = pd.read_csv(CG_DIR / "ITEM_COOC_TOPK_FALLBACK_TRAIN.csv", delimiter=",")
item_content= pd.read_csv(CG_DIR / "ITEM_CONTENT_NEIGHBORS_TRAIN.csv", delimiter=",")
```

```python
def normalize_cols(df):
    df.columns = [c.strip().lower() for c in df.columns]
    return df

# normalize column names
train_interactions = normalize_cols(train_interactions)
test_gt           = normalize_cols(test_gt)
df_cust           = normalize_cols(df_cust)
df_items          = normalize_cols(df_items)
persona_top       = normalize_cols(persona_top)
item_cooc         = normalize_cols(item_cooc)
item_content      = normalize_cols(item_content)

# dtypes
train_interactions["customer_key"] = train_interactions["customer_key"].astype(str)
train_interactions["product_id"]   = train_interactions["product_id"].astype(int)
train_interactions["strength"]     = pd.to_numeric(train_interactions["strength"], errors="coerce").fillna(0.0)

if "last_ts" in train_interactions.columns:
    train_interactions["last_ts"] = pd.to_datetime(train_interactions["last_ts"], errors="coerce")

df_cust["customer_key"] = df_cust["customer_key"].astype(str)
df_items["product_id"]  = df_items["product_id"].astype(int)

persona_top["product_id"] = persona_top["product_id"].astype(int)
item_cooc["product_id"]   = item_cooc["product_id"].astype(int)
item_cooc["neighbor_id"]  = item_cooc["neighbor_id"].astype(int)

item_content["product_id"]  = item_content["product_id"].astype(int)
item_content["neighbor_id"] = item_content["neighbor_id"].astype(int)

# strength_log
if "strength_log" not in train_interactions.columns:
    train_interactions["strength_log"] = np.log1p(train_interactions["strength"])
```

```python
import ast

def parse_gt_items_col(df, col="gt_items"):
    def _parse(x):
        # já é lista/array
        if isinstance(x, (list, tuple, np.ndarray)):
            return [int(i) for i in list(x)]
        # NaN escalar
        if x is None:
            return []
        if isinstance(x, float) and np.isnan(x):
            return []
        s = str(x).strip()
        if s == "" or s.lower() == "nan":
            return []
        # string tipo "[1, 2, 3]"
        try:
            v = ast.literal_eval(s)
            if isinstance(v, (list, tuple, np.ndarray)):
                return [int(i) for i in list(v)]
            # se vier "123"
            return [int(v)]
        except Exception:
            # fallback: extrair dígitos
            s2 = s.strip("[]")
            if s2 == "":
                return []
            return [int(z.strip()) for z in s2.split(",") if z.strip().isdigit()]
    df = df.copy()
    df[col] = df[col].apply(_parse)
    return df

test_gt = parse_gt_items_col(test_gt, "gt_items")
test_gt["customer_key"] = test_gt["customer_key"].astype(str)
```

```python
# maps úteis
persona_map = df_cust.drop_duplicates("customer_key").set_index("customer_key")["persona"].to_dict()

# persona_top: (persona -> list[(pid, score_pop)])
persona_top = persona_top.sort_values(["persona","rank"])
persona_top_dict = (persona_top.groupby("persona")
                    .apply(lambda d: list(zip(d["product_id"].astype(int), d["score_pop"].astype(float))))
                    .to_dict())

# cooc: (item -> list[(neighbor, cosine)])
item_cooc = item_cooc.sort_values(["product_id","rank"])
cooc_map = (item_cooc.groupby("product_id")
            .apply(lambda d: list(zip(d["neighbor_id"].astype(int), d["cosine"].astype(float))))
            .to_dict())

# content: (item -> list[(neighbor, score_from_rank, reason)])
item_content = item_content.sort_values(["product_id","rank"])
def _content_list(d):
    # score monotónico decrescente com rank (rank 1 melhor)
    score = 1.0 / (1.0 + d["rank"].astype(float).values)
    return list(zip(d["neighbor_id"].astype(int).values, score, d["reason"].astype(str).values))
content_map = item_content.groupby("product_id").apply(_content_list).to_dict()

# user history (set)
user_hist = train_interactions.groupby("customer_key")["product_id"].apply(lambda s: set(s.astype(int))).to_dict()

# global popularity fallback (train only)
pop_global = train_interactions.groupby("product_id")["strength_log"].sum().sort_values(ascending=False)
global_ranked = pop_global.index.astype(int).tolist()

def build_candidates_for_users(user_set,
                              TOP_PERSONA=100,
                              TOP_NEIGH=100,
                              GLOBAL_M=200,
                              filter_purchased=True):
    rows = []
    for u in user_set:
        persona = persona_map.get(u, None)
        hist = user_hist.get(u, set())

        # persona
        for pid, sc in persona_top_dict.get(persona, [])[:TOP_PERSONA]:
            if filter_purchased and pid in hist: 
                continue
            rows.append((u, pid, "persona_pop", float(sc)))

        # cooc
        for it in hist:
            for nid, sc in cooc_map.get(it, [])[:TOP_NEIGH]:
                if filter_purchased and nid in hist:
                    continue
                rows.append((u, nid, "cooc", float(sc)))

        # content
        for it in hist:
            for nid, sc, _reason in content_map.get(it, [])[:TOP_NEIGH]:
                if filter_purchased and nid in hist:
                    continue
                rows.append((u, nid, "content", float(sc)))

        # global fallback
        for pid in global_ranked[:GLOBAL_M]:
            if filter_purchased and pid in hist:
                continue
            rows.append((u, pid, "global_pop", float(pop_global.loc[pid])))

    cand = pd.DataFrame(rows, columns=["customer_key","product_id","source","raw_score"])

    # agregação (se o mesmo (u,item,source) aparece várias vezes)
    cand = (cand.groupby(["customer_key","product_id","source"], as_index=False)
                .agg(raw_score=("raw_score","max")))

    return cand

cand_all = build_candidates_for_users(set(test_gt["customer_key"]), TOP_PERSONA=100, TOP_NEIGH=100, GLOBAL_M=200)
print("cand_all:", cand_all.shape, "| users:", cand_all["customer_key"].nunique(), "| items:", cand_all["product_id"].nunique())
```

```python
users = test_gt["customer_key"].unique()
u2p = df_cust.drop_duplicates("customer_key").set_index("customer_key")["persona"].to_dict()

# estratifica por persona (quando possível)
df_u = pd.DataFrame({"customer_key": users})
df_u["persona"] = df_u["customer_key"].map(u2p).fillna("unknown")

train_users, val_users, test_users = [], [], []
for persona, g in df_u.groupby("persona"):
    arr = g["customer_key"].to_numpy()
    rng = np.random.RandomState(RANDOM_STATE)
    rng.shuffle(arr)
    n = len(arr)
    n_train = int(0.70*n)
    n_val   = int(0.15*n)
    train_users.extend(arr[:n_train])
    val_users.extend(arr[n_train:n_train+n_val])
    test_users.extend(arr[n_train+n_val:])

train_users = set(train_users)
val_users   = set(val_users)
test_users  = set(test_users)

print("[SPLIT USERS]", len(train_users), len(val_users), len(test_users))
```

```python
def minmax_by_user_source(df):
    df = df.copy()
    g = df.groupby(["customer_key","source"])["raw_score"]
    mn = g.transform("min")
    mx = g.transform("max")
    df["score"] = np.where(mx > mn, (df["raw_score"] - mn) / (mx - mn), 0.0)
    return df

def to_wide_features(cand):
    cand = minmax_by_user_source(cand)

    wide = (cand.pivot_table(index=["customer_key","product_id"],
                             columns="source",
                             values="score",
                             aggfunc="max",
                             fill_value=0.0)
              .reset_index())

    # garante colunas mesmo que alguma source falte
    for c in ["persona_pop","cooc","content","global_pop"]:
        if c not in wide.columns:
            wide[c] = 0.0

    # features agregadas “de apoio”
    wide["sum_source_scores"] = wide[["persona_pop","cooc","content","global_pop"]].sum(axis=1)
    wide["n_sources_hit"] = (wide[["persona_pop","cooc","content","global_pop"]] > 0).sum(axis=1)
    return wide

# item features (recomendado: calcular popularidade no treino, não no full)
item_train_agg = (train_interactions.groupby("product_id")
                  .agg(item_pop=("strength_log","sum"),
                       item_events=("product_id","size"),
                       item_users=("customer_key","nunique"))
                  .reset_index())

# catálogo (apenas atributos “estáticos”/não temporais)
cat_cols = [c for c in ["product_id","brand","hs_code","price_median"] if c in df_items.columns]
items_cat = df_items[cat_cols].copy()

# user features (apenas do treino)
train_max_ts = train_interactions["last_ts"].max() if "last_ts" in train_interactions.columns else None
u_agg = (train_interactions.groupby("customer_key")
         .agg(n_items=("product_id","nunique"),
              n_events=("product_id","size"),
              mon=("strength_log","sum"),
              avg_strength=("strength_log","mean"),
              last_ts=("last_ts","max") if "last_ts" in train_interactions.columns else ("customer_key","size"))
         .reset_index())
if train_max_ts is not None:
    u_agg["recency_days"] = (train_max_ts - u_agg["last_ts"]).dt.days.clip(lower=0)
else:
    u_agg["recency_days"] = 0

u_agg = u_agg.drop(columns=["last_ts"], errors="ignore")

# persona (categórica)
u_persona = df_cust.drop_duplicates("customer_key")[["customer_key","persona"]].copy()
u_persona["persona"] = u_persona["persona"].fillna("unknown")

# label building
gt_map = test_gt.set_index("customer_key")["gt_items"].to_dict()

def add_labels(wide):
    wide = wide.copy()
    wide["label"] = wide.apply(lambda r: 1 if r["product_id"] in gt_map.get(r["customer_key"], []) else 0, axis=1)
    return wide

# build wide features
wide_all = to_wide_features(cand_all)
wide_all = add_labels(wide_all)

# merge user/item features
wide_all = wide_all.merge(u_agg, on="customer_key", how="left")
wide_all = wide_all.merge(u_persona, on="customer_key", how="left")
wide_all = wide_all.merge(item_train_agg, on="product_id", how="left")
wide_all = wide_all.merge(items_cat, on="product_id", how="left")

# encode categoricals consistently
wide_all["persona_code"] = pd.factorize(wide_all["persona"].fillna("unknown"))[0]
wide_all["brand_code"]   = pd.factorize(wide_all.get("brand", pd.Series(["unknown"]*len(wide_all))).fillna("unknown"))[0]
wide_all["hs_code_code"] = pd.factorize(wide_all.get("hs_code", pd.Series(["unknown"]*len(wide_all))).fillna("unknown"))[0]

# fill missing numeric
for c in ["n_items","n_events","mon","avg_strength","recency_days","item_pop","item_events","item_users","price_median"]:
    if c in wide_all.columns:
        wide_all[c] = pd.to_numeric(wide_all[c], errors="coerce").fillna(0.0)

FEATURE_COLS = [
    "persona_pop","cooc","content","global_pop",
    "sum_source_scores","n_sources_hit",
    "item_pop","item_events","item_users","price_median",
    "n_items","n_events","mon","avg_strength","recency_days",
    "persona_code","brand_code","hs_code_code"
]

# split by users (no leakage across users)
def subset_by_users(df, user_set):
    return df[df["customer_key"].isin(user_set)].copy()

train_df = subset_by_users(wide_all, train_users)
val_df   = subset_by_users(wide_all, val_users)
test_df  = subset_by_users(wide_all, test_users)

print("train_df/val_df/test_df:", train_df.shape, val_df.shape, test_df.shape)

# (opcional mas recomendado) remover users sem positivos no treino (LTR precisa de sinal)
pos_by_u = train_df.groupby("customer_key")["label"].sum()
keep_train_users = set(pos_by_u[pos_by_u > 0].index)
train_df = train_df[train_df["customer_key"].isin(keep_train_users)].copy()
print("train users kept (pos>0):", train_df["customer_key"].nunique())

# LightGBM exige grupos contíguos por user
def make_X_y_group(df):
    df = df.sort_values("customer_key").reset_index(drop=True)
    X = df[FEATURE_COLS]
    y = df["label"].astype(int)
    group = df.groupby("customer_key").size().to_list()
    return df, X, y, group

train_df, X_train, y_train, g_train = make_X_y_group(train_df)
val_df,   X_val,   y_val,   g_val   = make_X_y_group(val_df)
test_df,  X_test,  y_test,  g_test  = make_X_y_group(test_df)

ranker = lgb.LGBMRanker(
    objective="lambdarank",
    metric="ndcg",
    learning_rate=0.05,
    n_estimators=2000,
    num_leaves=64,
    min_child_samples=20,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

ranker.fit(
    X_train, y_train, group=g_train,
    eval_set=[(X_val, y_val)],
    eval_group=[g_val],
    eval_at=[10],
    callbacks=[
        lgb.early_stopping(stopping_rounds=50),
        lgb.log_evaluation(period=50)
    ]
)

# save model
ranker.booster_.save_model(str(OUT_DIR / "lgbm_ranker.txt"))
```

```python
import math

def score_topk(df, model, K=10):
    df = df.copy()
    df["score_pred"] = model.predict(df[FEATURE_COLS])
    top = (df.sort_values(["customer_key","score_pred"], ascending=[True, False])
             .groupby("customer_key").head(K))
    recs = top.groupby("customer_key")["product_id"].apply(list).to_dict()
    return recs, df

def precision_at_k(recs, gt_map, k=10):
    vals=[]
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        vals.append(len([x for x in rec if x in gt]) / k if k>0 else 0.0)
    return float(np.mean(vals))

def recall_at_k(recs, gt_map, k=10):
    vals=[]
    for u, gt in gt_map.items():
        if len(gt)==0:
            vals.append(0.0); continue
        rec = recs.get(u, [])[:k]
        vals.append(len([x for x in rec if x in gt]) / len(gt))
    return float(np.mean(vals))

def hit_at_k(recs, gt_map, k=10):
    vals=[]
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        vals.append(1.0 if any(x in gt for x in rec) else 0.0)
    return float(np.mean(vals))

def ndcg_at_k(recs, gt_map, k=10):
    vals=[]
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        gains = [1 if r in gt else 0 for r in rec]
        dcg = sum([g / math.log2(i+2) for i, g in enumerate(gains)])
        ideal = [1]*min(len(gt), k)
        idcg = sum([g / math.log2(i+2) for i, g in enumerate(ideal)]) if len(ideal)>0 else 0.0
        vals.append(dcg/idcg if idcg>0 else 0.0)
    return float(np.mean(vals))

def map_at_k(recs, gt_map, k=10):
    vals=[]
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        if len(gt)==0:
            vals.append(0.0); continue
        hits=0; s=0.0
        for i, r in enumerate(rec, start=1):
            if r in gt:
                hits += 1
                s += hits/i
        denom = min(len(gt), k)
        vals.append(s/denom if denom>0 else 0.0)
    return float(np.mean(vals))

def candidate_recall(candidate_sets, gt_map):
    vals=[]
    for u, gt in gt_map.items():
        cand = candidate_sets.get(u, set())
        vals.append(1.0 if any(x in cand for x in gt) else 0.0)
    return float(np.mean(vals))

# build gt maps for each split (apenas users do split)
gt_map_all = gt_map
val_gt_map  = {u: gt_map_all.get(u, []) for u in val_df["customer_key"].unique()}
test_gt_map = {u: gt_map_all.get(u, []) for u in test_df["customer_key"].unique()}

# candidate recall
val_candsets  = val_df.groupby("customer_key")["product_id"].apply(set).to_dict()
test_candsets = test_df.groupby("customer_key")["product_id"].apply(set).to_dict()

val_recs,  val_scored  = score_topk(val_df, ranker, K=10)
test_recs, test_scored = score_topk(test_df, ranker, K=10)

print("\n[VAL] CandidateRecall:", candidate_recall(val_candsets, val_gt_map))
print("[VAL] Precision@10:", precision_at_k(val_recs, val_gt_map, 10))
print("[VAL] Recall@10   :", recall_at_k(val_recs, val_gt_map, 10))
print("[VAL] Hit@10      :", hit_at_k(val_recs, val_gt_map, 10))
print("[VAL] nDCG@10     :", ndcg_at_k(val_recs, val_gt_map, 10))
print("[VAL] MAP@10      :", map_at_k(val_recs, val_gt_map, 10))

print("\n[TEST] CandidateRecall:", candidate_recall(test_candsets, test_gt_map))
print("[TEST] Precision@10:", precision_at_k(test_recs, test_gt_map, 10))
print("[TEST] Recall@10   :", recall_at_k(test_recs, test_gt_map, 10))
print("[TEST] Hit@10      :", hit_at_k(test_recs, test_gt_map, 10))
print("[TEST] nDCG@10     :", ndcg_at_k(test_recs, test_gt_map, 10))
print("[TEST] MAP@10      :", map_at_k(test_recs, test_gt_map, 10))

# salvar recomendações
pd.DataFrame([{"customer_key":u, "recs":test_recs.get(u, [])} for u in test_recs]).to_csv(
    OUT_DIR / "reco_lgbm_test.csv", index=False, encoding="utf-8"
)
```

# Baseline for benchmark

```python


# 1. Preparação dos dados para Baselines 

# --- Baseline 1: Global Popularity ---
# Calcula popularidade baseada na soma de strength_log no TREINO
pop_global_series = train_interactions.groupby("product_id")["strength_log"].sum().sort_values(ascending=False)
global_ranked_items = pop_global_series.index.tolist()

# --- Baseline 2: Persona Popularity ---
# Merge com personas (usando df_cust que já tens carregado)
train_w_persona = train_interactions.merge(
    df_cust[["customer_key", "persona"]].drop_duplicates(), 
    on="customer_key", 
    how="left"
)
train_w_persona["persona"] = train_w_persona["persona"].fillna("unknown")

# Agrupa por persona e item
persona_pop_df = (train_w_persona.groupby(["persona", "product_id"])["strength_log"]
                  .sum()
                  .reset_index())

# Cria dicionário: {persona: [lista_ordenada_de_items]}
persona_ranked_map = (persona_pop_df.sort_values(["persona", "strength_log"], ascending=[True, False])
                      .groupby("persona")["product_id"]
                      .apply(list)
                      .to_dict())

# Mapa auxiliar de user -> persona
user_2_persona_map = df_cust.drop_duplicates("customer_key").set_index("customer_key")["persona"].to_dict()

# %%
# 2. Funções de Recomendação das Baselines
# ----------------------------------------

def recommend_global(user_id, k=10):
    """Retorna top-k itens globais não vistos no treino."""
    seen = user_hist.get(user_id, set())
    recs = []
    for item in global_ranked_items:
        if item not in seen:
            recs.append(item)
            if len(recs) == k:
                break
    return recs

def recommend_persona(user_id, k=10):
    """Retorna top-k itens da persona. Fallback para global se persona vazia."""
    seen = user_hist.get(user_id, set())
    persona = user_2_persona_map.get(user_id, "unknown")
    
    # Busca itens da persona, se não existir ou lista vazia, usa global
    candidate_items = persona_ranked_map.get(persona, [])
    
    # Se a persona não tiver items suficientes ou for desconhecida, completamos com global
    if not candidate_items:
        candidate_items = global_ranked_items
    
    recs = []
    # 1. Tenta preencher com itens da persona
    for item in candidate_items:
        if item not in seen:
            recs.append(item)
            if len(recs) == k:
                return recs
    
    # 2. Fallback: Se ainda não encheu K (e usámos items de persona), completa com global
    if len(recs) < k:
        for item in global_ranked_items:
            if item not in seen and item not in recs:
                recs.append(item)
                if len(recs) == k:
                    break
    return recs

# 3. Executar Previsões no TEST SET

# Importante: Avaliar EXATAMENTE nos mesmos users que o LightGBM (test_users)
# e usar o mesmo Ground Truth (test_gt_map)

print(f"Generating Baseline Recs for {len(test_users)} test users...")

recs_global_test = {}
recs_persona_test = {}

for u in test_users:
    recs_global_test[u] = recommend_global(u, k=10)
    recs_persona_test[u] = recommend_persona(u, k=10)


#redifinição de funções de avaliação

def precision_at_k(recs, gt_map, k=10):
    vals = []
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        vals.append(len([x for x in rec if x in gt]) / k if k > 0 else 0.0)
    return float(np.mean(vals)) if vals else 0.0

def recall_at_k(recs, gt_map, k=10):
    vals = []
    for u, gt in gt_map.items():
        if len(gt) == 0:
            vals.append(0.0)
            continue
        rec = recs.get(u, [])[:k]
        vals.append(len([x for x in rec if x in gt]) / len(gt))
    return float(np.mean(vals)) if vals else 0.0

def hit_at_k(recs, gt_map, k=10):
    vals = []
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        vals.append(1.0 if any(x in gt for x in rec) else 0.0)
    return float(np.mean(vals)) if vals else 0.0

def ndcg_at_k(recs, gt_map, k=10):
    vals = []
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        gains = [1 if r in gt else 0 for r in rec]
        # DCG
        dcg = sum([g / math.log2(i+2) for i, g in enumerate(gains)])
        # IDCG
        ideal = [1] * min(len(gt), k)
        idcg = sum([g / math.log2(i+2) for i, g in enumerate(ideal)]) if len(ideal) > 0 else 0.0
        
        vals.append(dcg / idcg if idcg > 0 else 0.0)
    return float(np.mean(vals)) if vals else 0.0

def map_at_k(recs, gt_map, k=10):
    vals = []
    for u, gt in gt_map.items():
        rec = recs.get(u, [])[:k]
        if len(gt) == 0:
            vals.append(0.0)
            continue
        hits = 0
        s = 0.0
        for i, r in enumerate(rec, start=1):
            if r in gt:
                hits += 1
                s += hits / i
        denom = min(len(gt), k)
        vals.append(s / denom if denom > 0 else 0.0)
    return float(np.mean(vals)) if vals else 0.0


def evaluate_model(recs_dict, gt_map, model_name="Model"):
    print(f"--- {model_name} ---")
    p10 = precision_at_k(recs_dict, gt_map, 10)
    r10 = recall_at_k(recs_dict, gt_map, 10)
    h10 = hit_at_k(recs_dict, gt_map, 10)
    n10 = ndcg_at_k(recs_dict, gt_map, 10)
    m10 = map_at_k(recs_dict, gt_map, 10)
    
    print(f"Precision@10: {p10:.4f}")
    print(f"Recall@10   : {r10:.4f}")
    print(f"Hit@10      : {h10:.4f}")
    print(f"nDCG@10     : {n10:.4f}")
    print(f"MAP@10      : {m10:.4f}")
    
    return {"Model": model_name, "P@10": p10, "R@10": r10, "Hit@10": h10, "nDCG@10": n10, "MAP@10": m10}

# Garante que as variáveis de input existem (do código anterior)
# test_recs, recs_global_test, recs_persona_test, test_gt_map, test_users

test_gt_map_filtered = {u: test_gt_map[u] for u in test_users if u in test_gt_map}

# Resultados
res_lgbm    = evaluate_model(test_recs, test_gt_map_filtered, model_name="LightGBM")
res_global  = evaluate_model(recs_global_test, test_gt_map_filtered, model_name="Baseline Global")
res_persona = evaluate_model(recs_persona_test, test_gt_map_filtered, model_name="Baseline Persona")

df_results = pd.DataFrame([res_global, res_persona, res_lgbm])
df_results = df_results.set_index("Model")
print("\n=== FINAL BENCHMARK (TEST SET) ===")
print(df_results)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Função auxiliar para calcular métrica por grupo
def get_metrics_by_persona(recs_dict, gt_map, df_pers, model_name):
    # Merge users com personas
    users_eval = list(recs_dict.keys())
    temp_df = pd.DataFrame({"customer_key": users_eval})
    temp_df = temp_df.merge(df_pers, on="customer_key", how="left").fillna("Unknown")
    
    results = []
    for persona, group in temp_df.groupby("persona"):
        users_in_persona = group["customer_key"].unique()
        # Filtrar o mapa de GT e Recs para esta persona
        gt_subset = {u: gt_map[u] for u in users_in_persona if u in gt_map}
        
        score = ndcg_at_k(recs_dict, gt_subset, k=10)
        results.append({"Persona": persona, "nDCG@10": score, "Model": model_name})
    
    return pd.DataFrame(results)

# 3. Calcular para os 3 modelos
df_metric_global  = get_metrics_by_persona(recs_global_test, test_gt_map, df_cust, "Global Pop")
df_metric_persona = get_metrics_by_persona(recs_persona_test, test_gt_map, df_cust, "Persona Pop")
df_metric_lgbm    = get_metrics_by_persona(test_recs, test_gt_map, df_cust, "LightGBM")

# 4. Juntar e Visualizar
df_final_persona = pd.concat([df_metric_global, df_metric_persona, df_metric_lgbm])

plt.figure(figsize=(12, 6))
sns.barplot(data=df_final_persona, x="Persona", y="nDCG@10", hue="Model", palette="viridis")
plt.title("Performance of model per Persona (nDCG@10)")
plt.ylabel("nDCG@10")
plt.xticks(rotation=45)
plt.gcf().set_facecolor('#fff7eb')
plt.gca().set_facecolor('#fff7eb')
plt.tight_layout()
plt.show()
```
