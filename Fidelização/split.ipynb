# Split

```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path(r"CSVTables\Recommender")
OUT_DIR  = Path(r"CSVTables\Recommender\Split_TrainTest")
OUT_DIR.mkdir(parents=True, exist_ok=True)

# --- load
interactions = pd.read_csv(DATA_DIR / "INTERACTIONS.csv", delimiter=",")

# --- normalize dtypes
interactions["customer_key"] = interactions["customer_key"].astype(str)
interactions["product_id"]   = pd.to_numeric(interactions["product_id"], errors="coerce")
interactions["strength"]     = pd.to_numeric(interactions["strength"], errors="coerce").fillna(0.0)
interactions["last_ts"]      = pd.to_datetime(interactions["last_ts"], errors="coerce")

interactions = interactions.dropna(subset=["customer_key","product_id","last_ts"]).copy()
interactions["product_id"] = interactions["product_id"].astype(int)

# sort temporal
interactions = interactions.sort_values(["customer_key","last_ts"]).reset_index(drop=True)

# --- elegibilidade: >=2 timestamps (preferível) senão >=2 interações
n_ts = interactions.groupby("customer_key")["last_ts"].nunique()
users_eval = n_ts[n_ts >= 2].index
inter_eval = interactions[interactions["customer_key"].isin(users_eval)].copy()

print("Total users:", interactions["customer_key"].nunique())
print("Users eligible (>=2 timestamps):", len(users_eval))
print("Eligible interactions:", inter_eval.shape)

# --- último timestamp por user
last_ts_per_user = (
    inter_eval.groupby("customer_key")["last_ts"].max()
    .rename("test_ts")
    .reset_index()
)

tmp = inter_eval.merge(last_ts_per_user, on="customer_key", how="left")

# test rows = tudo no último timestamp (aproxima “último basket”)
test_rows = tmp[tmp["last_ts"] == tmp["test_ts"]].copy()

def to_unique_list(s):
    return sorted(set(s.astype(int).tolist()))

test_gt = (
    test_rows.groupby("customer_key")["product_id"]
    .apply(to_unique_list)
    .rename("gt_items")
    .reset_index()
)

# train = remove essas linhas de teste
# (remove exatamente as linhas customer+product+timestamp do test)
mark = test_rows[["customer_key","product_id","last_ts"]].drop_duplicates().assign(_mark=1)

tmp2 = tmp.merge(mark, on=["customer_key","product_id","last_ts"], how="left")
tmp2["_mark"] = tmp2["_mark"].fillna(0).astype(int)

train_interactions = tmp2[tmp2["_mark"] == 0].drop(columns=["_mark"]).copy()

# útil: strength_log
train_interactions["strength_log"] = np.log1p(train_interactions["strength"])

# --- save (parquet recomendado)
train_interactions.to_parquet(OUT_DIR / "train_interactions.parquet", index=False)
test_gt.to_parquet(OUT_DIR / "test_gt.parquet", index=False)

# opcional: CSV
train_interactions.to_csv(OUT_DIR / "train_interactions.csv", index=False, encoding="utf-8")
test_gt.to_csv(OUT_DIR / "test_gt.csv", index=False, encoding="utf-8")

print("\n[SAVED]")
print("train_interactions:", train_interactions.shape, "| users:", train_interactions["customer_key"].nunique())
print("test_gt:", test_gt.shape, "| users:", test_gt["customer_key"].nunique())

# sanity: overlap
print("Coverage test users in train users:",
      (set(test_gt["customer_key"]) <= set(train_interactions["customer_key"])) )
```
