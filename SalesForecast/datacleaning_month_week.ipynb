Excluido Marketplace | Transactions > 2025-04-30 | Qty Invoiced == 0

Selecionamos SKUs vendidos em 2025 e com CatalogIds

```python
# Filtrar successful orders  
sales_order = sales_order[(sales_order['state']=='complete') | (sales_order['state']=='processing')]

# Excluir pedidos com AMZ ou EBAY no increment_id
sales_order = sales_order[~sales_order['increment_id'].str.contains('AMZ|EBAY', na=False)]

# Converter created_at para datetime e extrair data
sales_order['created_at'] = pd.to_datetime(sales_order['created_at'])
sales_order['date'] = sales_order['created_at'].dt.date

# Excluir datas depois de abril 2025
sales_order = sales_order[sales_order['created_at'] <= '2025-04-30']

sales_item = sales_item[sales_item['order_id'].isin(sales_order['entity_id'])]

print(f"Após remover marketplaces e datas fora do intervalo: {sales_order.shape[0]} orders | {sales_item.shape[0]} items")

## TIRAR ITEMS NAO INVOICED 
sales_item = sales_item[sales_item['qty_invoiced']!=0]

# CRIAR LISTA DE SKUs QUE VENDERAM EM 2025
# Juntar orders com items para encontrar os SKUs vendidos em 2025
sales_2025 = sales_order[sales_order['created_at'] >= '2025-01-01'][['entity_id']]
items_2025 = sales_item.merge(sales_2025, left_on='order_id', right_on='entity_id', how='inner')
skus_sold_2025 = items_2025['sku'].unique()
product_ids_sold_2025 = items_2025['product_id'].unique()

print(f"SKUs que venderam em 2025: {len(skus_sold_2025)}")

# Produtos disponíveis no catálogo (attribute_id = 96 value=1 ou 174 value=0) verifiquei com base nos tops SKUs do catalogo
available_products_96 = catalog_product_entity_int[(catalog_product_entity_int['ATTRIBUTE_ID']==96) & (catalog_product_entity_int['VALUE']==1)]['ENTITY_ID'].unique()
available_products_174 = catalog_product_entity_int[(catalog_product_entity_int['ATTRIBUTE_ID']==174) & (catalog_product_entity_int['VALUE']==0)]['ENTITY_ID'].unique()
available_products_catalog = set(available_products_96).union(set(available_products_174))

print(f"Produtos disponíveis no catálogo: {len(available_products_catalog)}")

# COMBINAR: Produtos disponíveis no catálogo OU que venderam em 2025
available_products = available_products_catalog.union(set(product_ids_sold_2025))

print(f"Total de produtos a considerar (catálogo + vendidos 2025): {len(available_products)}")

# Filtrar sales_item para incluir apenas produtos disponíveis ou que venderam em 2025
sales_item = sales_item[sales_item['product_id'].isin(available_products)]

print(f"\n Sales rows after marketplace cleaning: {sales_order.shape[0]} | Items rows after cleaning: {sales_item.shape[0]}")

# Só mostrar orders que têm items válidos
sales_order = sales_order[sales_order['entity_id'].isin(sales_item['order_id'].unique())]
print(f" Final: Sales rows: {sales_order.shape[0]} | Items rows: {sales_item.shape[0]}")
```

```python
# ADICIONAR HAS_CATALOG_PRODUCT_CATEGORY e CATEGORY 
"""
Extrai a categoria de um SKU seguindo o caminho:
1. SKU -> product_id (sales_item)
2. product_id -> entity_id (product_entity_int) WHERE attribute_id=168
3. value (option_id) -> option_id (eav_attribute_option_value) WHERE store_id=0
4. Retorna value (nome da categoria)
"""

# 0. REMOVER COLUNAS ANTIGAS (se já existirem) - EVITA ERRO EM RERUN
if 'CATEGORY' in df_input.columns:
    df_input = df_input.drop(columns=['CATEGORY'])
if 'HAS_CATALOG_PRODUCT_CATEGORY' in df_input.columns:
    df_input = df_input.drop(columns=['HAS_CATALOG_PRODUCT_CATEGORY'])
    
# 1. Criar lookup table de product_id -> option_id (attribute_id=168)
product_to_option = catalog_product_entity_int[
    catalog_product_entity_int['ATTRIBUTE_ID'] == 168
][['ENTITY_ID', 'VALUE']].copy()

product_to_option.columns = ['product_id', 'option_id']

# 2. Criar lookup table de option_id -> category_name (store_id=0)
option_to_category = eav_attribute_option_value[
    eav_attribute_option_value['store_id'] == 0
][['option_id', 'value']].copy()

option_to_category.columns = ['option_id', 'category_name']

# 3. Criar mapeamento completo: product_id -> category_name
product_category_map = product_to_option.merge(
    option_to_category, 
    on='option_id', 
    how='left'
)[['product_id', 'category_name']]

# 4. Aplicar mapeamento ao df_input via merge (muito mais rápido que .map())
df_input = df_input.merge(
    product_category_map, 
    on='product_id', 
    how='left'
)

# 5. Renomear e criar coluna HAS_CATALOG_PRODUCT_CATEGORY
df_input.rename(columns={'category_name': 'CATEGORY'}, inplace=True)
df_input['HAS_CATALOG_PRODUCT_CATEGORY'] = df_input['CATEGORY'].notna().astype(int)

#Exclude "gift" and "influencer" skus
df_input = df_input[~df_input['sku'].str.contains('gift|influencer', case=False, na=False)]

# 6. Preencher valores nulos com 0
df_input['CATEGORY'] = df_input['CATEGORY'].fillna(0)

# 7. Estatísticas
print(f"\n=== Mapeamento de Categorias Concluído ===")
print(f"Produtos únicos no df_input: {df_input['product_id'].nunique()}")
print(f"SKUs únicos no df_input: {df_input['sku'].nunique()}")
print(f"Linhas com categoria: {df_input['HAS_CATALOG_PRODUCT_CATEGORY'].sum()}")
print(f"Linhas sem categoria: {(df_input['HAS_CATALOG_PRODUCT_CATEGORY'] == 0).sum()}")

print("\n=== Distribuição de Categorias ===")
print(len(df_input['CATEGORY'].unique()), "categorias únicas encontradas:")
pd.set_option('display.max_rows', None)
print(df_input['CATEGORY'].value_counts())

print("\n=== Primeiras linhas com categorias ===")
print(df_input[['sku', 'product_id', 'CATEGORY', 'HAS_CATALOG_PRODUCT_CATEGORY']].head(20))


```

```python

# Agregar por SKU
sku_analysis = merged_data[~merged_data['sku'].str.contains('gift|influencer', case=False, na=False)].groupby('sku').agg({
    'sku_revenue': 'sum',    # Revenue total por SKU (usando fórmula correta)
    'qty_invoiced': 'sum'      # Número de vendas
}).reset_index()

sku_analysis.columns = ['sku', 'revenue', 'num_sales']

# Calcular variabilidade (coeficiente de variação) por SKU
# Agrupar vendas por SKU e MÊS (não por dia)
sku_monthly_sales = merged_data.groupby(['sku', 'month'])['qty_invoiced'].sum().reset_index()

# NOVO: Para cada SKU, encontrar mês da primeira e última venda
sku_date_range = sku_monthly_sales.groupby('sku')['month'].agg(['min', 'max']).reset_index()
sku_date_range.columns = ['sku', 'first_sale_month', 'last_sale_month']

print(f"Grid considerando apenas meses entre primeira e última venda de cada SKU...")
print(f"Exemplo de ranges por SKU:")
print(sku_date_range.head(10))

# Para cada SKU, criar range de meses entre primeira e última venda
sku_list = []
for _, row in sku_date_range.iterrows():
    sku = row['sku']
    first_month = row['first_sale_month']
    last_month = row['last_sale_month']
    
    # Criar range de meses para este SKU específico
    sku_months = pd.period_range(start=first_month, end=last_month, freq='M')
    
    # Criar DataFrame com todas as combinações deste SKU
    sku_df = pd.DataFrame({
        'sku': sku,
        'month': sku_months
    })
    
    sku_list.append(sku_df)

# Combinar todos os SKUs
sku_monthly_complete = pd.concat(sku_list, ignore_index=True)

# Merge com os dados reais, preenchendo com 0 onde não há vendas
sku_monthly_complete = sku_monthly_complete.merge(
    sku_monthly_sales, 
    on=['sku', 'month'], 
    how='left'
).fillna(0)

print(f"\nGrid completo criado: {len(sku_monthly_complete)} linhas (SKU x mês relevante)")
print(f"\nExemplo de SKU com meses completos (apenas entre primeira e última venda):")
print(sku_monthly_complete[sku_monthly_complete['sku'] == '000266EV'])

# Calcular estatísticas (mean, std, cv) com os zeros incluídos
sku_std = sku_monthly_complete.groupby('sku')['qty_invoiced'].agg(['mean', 'std']).reset_index()
sku_std['cv'] = (sku_std['std'] / sku_std['mean']).fillna(0)

print(f"\nEstatísticas com zeros incluídos (apenas entre primeira e última venda):")
print(sku_std.head())

# Juntar com análise principal
sku_analysis = sku_analysis.merge(sku_std[['sku', 'cv']], on='sku', how='left')
# print(sku_analysis.head())
# Classificação ABC 

sku_analysis = sku_analysis.sort_values('num_sales', ascending=False)
sku_analysis['numsales_cum_pct'] = (sku_analysis['num_sales'].cumsum() / sku_analysis['num_sales'].sum()) * 100
#print(sku_analysis.head())

sku_analysis['ABC'] = 'C'
sku_analysis.loc[sku_analysis['numsales_cum_pct'] <= 40, 'ABC'] = 'A' #  80% volume sao A
sku_analysis.loc[(sku_analysis['numsales_cum_pct'] > 40) & (sku_analysis['numsales_cum_pct'] <= 70), 'ABC'] = 'B' # 80-95% volume sao B
# Classificação XYZ (baseada em coeficiente de variação)
sku_analysis = sku_analysis.sort_values('cv')
sku_analysis['XYZ'] = 'Z'
"""
sku_analysis.loc[sku_analysis['cv'] <= sku_analysis['cv'].quantile(0.33), 'XYZ'] = 'X' # 1º 33% sao X ( interpolacao linear (Max-Min)*0.33 + Min)
sku_analysis.loc[(sku_analysis['cv'] > sku_analysis['cv'].quantile(0.33)) & # 33%-67% sao Y
                 (sku_analysis['cv'] <= sku_analysis['cv'].quantile(0.67)), 'XYZ'] = 'Y'
"""

sku_analysis.loc[sku_analysis['cv'] <= 0.5, 'XYZ'] = 'X' # 1º 33% sao X ( interpolacao linear (Max-Min)*0.33 + Min)
sku_analysis.loc[(sku_analysis['cv'] > 0.5) & # 33%-67% sao Y
                 (sku_analysis['cv'] <= 1), 'XYZ'] = 'Y'

# Combinar classificações
sku_analysis['ABC_XYZ'] = sku_analysis['ABC'] + sku_analysis['XYZ']

# Mostrar resumo
print(f'\nTotal de SKUs analisados: {len(sku_analysis)}')
print(f'\nDistribuição ABC:')
print(sku_analysis['ABC'].value_counts().sort_index())
print(f'\nDistribuição XYZ:')
print(sku_analysis['XYZ'].value_counts().sort_index())
print(f'\nDistribuição ABC-XYZ:')
print(sku_analysis['ABC_XYZ'].value_counts())

sku_analysis.head(20)
```

# MODELO LSTM PARA PREVISÃO DE VENDAS (COM SEQUÊNCIAS TEMPORAIS)

**Períodos de Dados:**
- **Treino**: 2023-01-01 → 2024-12-31
- **Validação**: 2025-01-01 → 2025-03-31  
- **Teste**: 2025-04-01 → 2025-04-30

**Arquitetura Sequencial:**
- **Timesteps**: 1 mes de histórico
- **Objetivo**: Prever vendas da próxima semana usando padrão do 1 mes

**Features de Entrada (por timestep):**
- Numéricas: MONTH_SIN, MONTH_COS, PRICE_PER_UNIT, LAG_SALES_1W_ANO_ANTERIOR, STORE_DISCOUNT_RATE, COUPON_DISCOUNT_RATE
- Binárias: HAS_LAG_SALES, IS_BLACKFRIDAY, IS_SUMMER, HAS_COUPON, HAS_CATALOG_PRODUCT_CATEGORY
- Categóricas (Embeddings): SKU, CATEGORY

**Output:**
- QTY (quantidade vendida no próximo mês)

```python
# ================================================================================
# PREPARAÇÃO DOS DADOS PARA MODELO LSTM COM SEQUÊNCIAS TEMPORAIS
# ================================================================================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.preprocessing import LabelEncoder, StandardScaler
import warnings
warnings.filterwarnings('ignore')

print("TensorFlow version:", tf.__version__)
print("Keras version:", keras.__version__)

# Criar cópia do dataframe para processamento
df_model = df_input.copy()

"""
# Converter week_year para datetime (primeira segunda-feira da semana)
df_model['date'] = pd.to_datetime(
    df_model['week_year'].str.replace('_W', '-W') + '-1',
    format='%Y-W%W-%w'
)
"""

df_model['date'] = pd.to_datetime(
    df_model['year_month'].str.replace('_M', '-') + '-01',
    format='%Y-%m-%d'
)

# ORDENAR POR SKU E DATA (ESSENCIAL PARA SEQUÊNCIAS)
df_model = df_model.sort_values(['sku', 'date']).reset_index(drop=True)

print("\n=== Dataset Temporal ===")
print(f"Período total: {df_model['date'].min()} → {df_model['date'].max()}")
print(f"Total de registros: {len(df_model)}")
print(f"SKUs únicos: {df_model['sku'].nunique()}")

# Split temporal usando week_year
train_mask = (df_model['date'] >= '2023-01-01') & (df_model['date'] <= '2024-12-31')
val_mask = (df_model['date'] >= '2024-01-01') & (df_model['date'] <= '2025-03-31')
test_mask = (df_model['date'] >= '2025-04-01') & (df_model['date'] <= '2025-04-30')

df_train = df_model[train_mask].copy()
df_val = df_model[val_mask].copy()
df_test = df_model[test_mask].copy()

print("\n=== Split Temporal ===")
print(f"Treino:    {len(df_train):6d} registros | {df_train['date'].min()} → {df_train['date'].max()}")
print(f"Validação: {len(df_val):6d} registros | {df_val['date'].min()} → {df_val['date'].max()}")
print(f"Teste:     {len(df_test):6d} registros | {df_test['date'].min()} → {df_test['date'].max()}")

# Distribuição de SKUs
print("\n=== Distribuição de SKUs ===")
print(f"SKUs únicos no treino:    {df_train['sku'].nunique()}")
print(f"SKUs únicos na validação: {df_val['sku'].nunique()}")
print(f"SKUs únicos no teste:     {df_test['sku'].nunique()}")

# Verificar cold start (SKUs sem histórico)
train_skus = set(df_train['sku'].unique())
val_skus = set(df_val['sku'].unique())
test_skus = set(df_test['sku'].unique())

cold_start_val = val_skus - train_skus
cold_start_test = test_skus - train_skus

if len(cold_start_val) > 0:
    print(f"\n SKUs em validação sem histórico de treino: {len(cold_start_val)}")
if len(cold_start_test) > 0:
    print(f" SKUs em teste sem histórico de treino: {len(cold_start_test)}")

print("\n✓ Dados ordenados e prontos para criação de sequências temporais")

#df_model
```

Embeddings para SKUs e Categorias

```python
# ================================================================================
# ENCODING DE VARIÁVEIS CATEGÓRICAS E NORMALIZAÇÃO
# ================================================================================

# Definir features numéricas e binárias
"""
numeric_features = ['WEEK_SIN', 'WEEK_COS', 'PRICE_PER_UNIT', 'LAG_SALES_1W_ANO_ANTERIOR', 
                    'STORE_DISCOUNT_RATE', 'COUPON_DISCOUNT_RATE']
"""
numeric_features = ['MONTH_SIN', 'MONTH_COS', 'PRICE_PER_UNIT', 'LAG_SALES_1M_ANO_ANTERIOR',
                    'STORE_DISCOUNT_RATE', 'COUPON_DISCOUNT_RATE']
binary_features = ['HAS_LAG_SALES', 'IS_BLACKFRIDAY', 'IS_SUMMER', 'HAS_COUPON', 
                   'HAS_CATALOG_PRODUCT_CATEGORY']

# Verificar quais features existem
available_numeric = [f for f in numeric_features if f in df_train.columns]
available_binary = [f for f in binary_features if f in df_train.columns]

print("=== Features Disponíveis ===")
print(f"Numéricas ({len(available_numeric)}): {available_numeric}")
print(f"Binárias ({len(available_binary)}): {available_binary}")

# 1. TRATAR VALORES INFINITOS E NaN
for df in [df_train, df_val, df_test]:
    for col in available_numeric:
        # Substituir inf por NaN e depois preencher com 0
        df[col] = df[col].replace([np.inf, -np.inf], np.nan).fillna(0)

print("\n✓ Valores infinitos tratados")

# 2. ENCODING DE VARIÁVEIS CATEGÓRICAS
sku_encoder = LabelEncoder()
category_encoder = LabelEncoder()

# Converter CATEGORY para string
for df in [df_train, df_val, df_test]:
    df['CATEGORY'] = df['CATEGORY'].astype(str)

# Fit apenas no treino
sku_encoder.fit(df_train['sku'])
category_encoder.fit(df_train['CATEGORY'])

# Helper function para encoding com valores desconhecidos
def encode_with_unknown(data, encoder, unknown_idx):
    """Encode data usando encoder, mapeando valores desconhecidos para unknown_idx"""
    return data.map(lambda x: encoder.transform([x])[0] if x in encoder.classes_ else unknown_idx)

# Índices para valores desconhecidos
unknown_sku_idx = len(sku_encoder.classes_)
unknown_cat_idx = len(category_encoder.classes_)

# Transform em todos os conjuntos
df_train['sku_encoded'] = sku_encoder.transform(df_train['sku'])
df_train['category_encoded'] = category_encoder.transform(df_train['CATEGORY'])

df_val['sku_encoded'] = encode_with_unknown(df_val['sku'], sku_encoder, unknown_sku_idx)
df_val['category_encoded'] = encode_with_unknown(df_val['CATEGORY'], category_encoder, unknown_cat_idx)

df_test['sku_encoded'] = encode_with_unknown(df_test['sku'], sku_encoder, unknown_sku_idx)
df_test['category_encoded'] = encode_with_unknown(df_test['CATEGORY'], category_encoder, unknown_cat_idx)

print("\n=== Encoding Completado ===")
print(f"SKUs únicos:       {len(sku_encoder.classes_)}")
print(f"Categorias únicas: {len(category_encoder.classes_)}")
print(f"SKUs desconhecidos (val): {(df_val['sku_encoded'] == unknown_sku_idx).sum()}")
print(f"SKUs desconhecidos (test): {(df_test['sku_encoded'] == unknown_sku_idx).sum()}")
print(f"Categorias desconhecidas (val): {(df_val['category_encoded'] == unknown_cat_idx).sum()}")
print(f"Categorias desconhecidas (test): {(df_test['category_encoded'] == unknown_cat_idx).sum()}")

# 3. NORMALIZAÇÃO DE FEATURES NUMÉRICAS
scaler = StandardScaler()

# Fit e transform no treino
df_train.loc[:, available_numeric] = scaler.fit_transform(df_train[available_numeric])

# Transform em val e test
df_val.loc[:, available_numeric] = scaler.transform(df_val[available_numeric])
df_test.loc[:, available_numeric] = scaler.transform(df_test[available_numeric])

print("\n✓ Normalização completada")

# 4. RESUMO FINAL
print("\n=== Target (QTY) ===")
print(f"Treino    - Média: {df_train['QTY'].mean():.2f} | Std: {df_train['QTY'].std():.2f} | Max: {df_train['QTY'].max():.0f}")
print(f"Validação - Média: {df_val['QTY'].mean():.2f} | Std: {df_val['QTY'].std():.2f} | Max: {df_val['QTY'].max():.0f}")
print(f"Teste     - Média: {df_test['QTY'].mean():.2f} | Std: {df_test['QTY'].std():.2f} | Max: {df_test['QTY'].max():.0f}")

# Dimensões para embeddings (incluindo classe "unknown")
num_skus = len(sku_encoder.classes_) + 1
num_categories = len(category_encoder.classes_) + 1

print(f"\n=== Dimensões para Embeddings ===")
print(f"Vocabulário SKU: {num_skus}")
print(f"Vocabulário CATEGORY: {num_categories}")
```

```python
# ================================================================================
# MODELO LSTM SIMPLIFICADO (PARA POUCOS DADOS)
# ================================================================================

if 'num_skus' not in locals():
    print("❌ Execute a célula de encoding primeiro")
else:
    # Dimensões de embedding
    sku_embedding_dim = 8  
    category_embedding_dim = 4 
    
    # Número de timesteps e features
    TIMESTEPS = 1
    num_numeric_features = len(available_numeric) + len(available_binary)
    
    print("=== Arquitetura LSTM SIMPLIFICADA ===")
    print(f"Timesteps: {TIMESTEPS}")
    print(f"Features por timestep: {num_numeric_features}")
    print(f"SKU Embedding: {num_skus} → {sku_embedding_dim} dimensões ")
    print(f"Category Embedding: {num_categories} → {category_embedding_dim} dimensões ")
    
    # ===== INPUTS =====
    input_numeric = layers.Input(shape=(TIMESTEPS, num_numeric_features), name='numeric_sequences')
    input_sku = layers.Input(shape=(1,), name='sku_id')
    input_category = layers.Input(shape=(1,), name='category_id')
    
    # ===== EMBEDDINGS (REDUZIDOS) =====
    sku_embedding = layers.Embedding(
        input_dim=num_skus,
        output_dim=sku_embedding_dim,
        name='sku_embedding'
    )(input_sku)
    sku_embedding = layers.Flatten()(sku_embedding)
    
    category_embedding = layers.Embedding(
        input_dim=num_categories,
        output_dim=category_embedding_dim,
        name='category_embedding'
    )(input_category)
    category_embedding = layers.Flatten()(category_embedding)
    
    # ===== LSTM SIMPLIFICADO (1 CAMADA APENAS) =====
    # 1 camada LSTM(32)
    x = layers.LSTM(32, return_sequences=False, name='lstm_1')(input_numeric)
    x = layers.Dropout(0.2)(x)  # Dropout reduzido de 0.3 para 0.2
    
    # ===== CONCATENAR COM EMBEDDINGS =====
    combined = layers.Concatenate()([x, sku_embedding, category_embedding])
    
    # ===== CAMADA DENSA SIMPLIFICADA =====
    # Dense(16) 
    x = layers.Dense(16, activation='relu', name='dense_1')(combined)
    x = layers.Dropout(0.1)(x)  
    
    # ===== OUTPUT =====
    output = layers.Dense(1, activation='relu', name='output_qty')(x)
    
    # ===== CRIAR MODELO =====
    model = keras.Model(
        inputs=[input_numeric, input_sku, input_category],
        outputs=output,
        name='LSTM_Sales_Forecasting_Simplified'
    )
    
    # ===== COMPILAR COM REGULARIZAÇÃO =====
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.0005),  # Learning rate menor
        loss='mse',
        metrics=['mae', 'mape']
    )
    
    # ===== RESUMO DO MODELO =====
    print("\n" + "="*80)
    model.summary()
    print("="*80)
    
    total_params = model.count_params()
```

```python
# ================================================================================
# CRIAR SEQUÊNCIAS TEMPORAIS PARA LSTM 
# ================================================================================

def create_sequences(df, all_features, timesteps=4):
    """
    Cria sequências temporais para cada SKU de forma otimizada.
    
    Args:
        df: DataFrame ordenado por sku e date
        all_features: Lista de features numéricas/binárias
        timesteps: Número de semanas no histórico (janela temporal)
    
    Returns:
        X_numeric: Array (samples, timesteps, num_features)
        X_sku: Array (samples,) - SKU IDs
        X_category: Array (samples,) - Category IDs
        y: Array (samples,) - Target
    """
    num_features = len(all_features)
    
    # Pré-calcular número total de sequências para pré-alocar arrays
    total_sequences = 0
    sku_groups = df.groupby('sku', sort=False)
    for _, group in sku_groups:
        n = len(group)
        if n >= timesteps + 1:
            total_sequences += n - timesteps
    
    # Pré-alocar arrays (muito mais rápido que append)
    sequences_numeric = np.empty((total_sequences, timesteps, num_features), dtype=np.float32)
    sequences_sku = np.empty(total_sequences, dtype=np.int32)
    sequences_category = np.empty(total_sequences, dtype=np.int32)
    targets = np.empty(total_sequences, dtype=np.float32)
    
    idx = 0
    
    # Processar cada SKU usando groupby (mais rápido que unique + filtragem)
    for sku, sku_data in sku_groups:
        n = len(sku_data)
        
        # Pular SKUs com poucos dados
        if n < timesteps + 1:
            continue
        
        # Converter para numpy arrays uma vez (evita conversões repetidas)
        features_array = sku_data[all_features].to_numpy(dtype=np.float32)
        sku_encoded_array = sku_data['sku_encoded'].to_numpy(dtype=np.int32)
        category_encoded_array = sku_data['category_encoded'].to_numpy(dtype=np.int32)
        qty_array = sku_data['QTY'].to_numpy(dtype=np.float32)
        
        # Criar todas as sequências deste SKU usando slicing vetorizado
        num_sequences = n - timesteps
        
        for i in range(num_sequences):
            # Janela de timesteps (slicing é muito rápido em numpy)
            sequences_numeric[idx] = features_array[i:i+timesteps]
            
            # SKU e Category do último timestep da janela
            sequences_sku[idx] = sku_encoded_array[i+timesteps-1]
            sequences_category[idx] = category_encoded_array[i+timesteps-1]
            
            # Target (próxima semana)
            targets[idx] = qty_array[i+timesteps]
            
            idx += 1
    
    return sequences_numeric, sequences_sku, sequences_category, targets

if 'model' not in locals():
    print(" Execute a célula de construção do modelo primeiro")
else:
    # Definir número de timesteps (semanas de histórico)
    TIMESTEPS = 1  # Usar últimas 4 semanas para prever a próxima
    
    # Combinar features numéricas + binárias
    all_features = available_numeric + available_binary
    
    print(f"=== Sequências Temporais ===")
    print(f"Timesteps: {TIMESTEPS} semanas/mes")
    print(f"Features por timestep: {len(all_features)}")
    print(f"Features: {all_features}\n")
    
    # Criar sequências para cada conjunto
    import time
    
    X_train_numeric, X_train_sku, X_train_category, y_train = create_sequences(
        df_train, all_features, TIMESTEPS
    )
    
    X_val_numeric, X_val_sku, X_val_category, y_val = create_sequences(
        df_val, all_features, TIMESTEPS
    )
    
    X_test_numeric, X_test_sku, X_test_category, y_test = create_sequences(
        df_test, all_features, TIMESTEPS
    )
    
    print("\n=== Sequências Criadas ===")
    print(f"\nTreino:")
    print(f"  Numeric sequences: {X_train_numeric.shape} (samples, timesteps, features)")
    print(f"  SKU IDs:          {X_train_sku.shape}")
    print(f"  Category IDs:     {X_train_category.shape}")
    print(f"  Target (QTY):     {y_train.shape}")
    
    print(f"\nValidação:")
    print(f"  Numeric sequences: {X_val_numeric.shape} (samples, timesteps, features)")
    print(f"  SKU IDs:          {X_val_sku.shape}")
    print(f"  Category IDs:     {X_val_category.shape}")
    print(f"  Target (QTY):     {y_val.shape}")
    
    print(f"\nTeste:")
    print(f"  Numeric sequences: {X_test_numeric.shape} (samples, timesteps, features)")
    print(f"  SKU IDs:          {X_test_sku.shape}")
    print(f"  Category IDs:     {X_test_category.shape}")
    print(f"  Target (QTY):     {y_test.shape}")
    
    print(f"\n✓ Dados sequenciais prontos para LSTM!")
    print(f"✓ Total de sequências: {len(y_train) + len(y_val) + len(y_test)}")
```

```python
# ================================================================================
# COMPARAÇÃO MODELO LSTM vs BASELINES PARA TODOS OS SKUs
# ================================================================================

import warnings
warnings.filterwarnings('ignore')

# SKUs a avaliar (todas as categorias AB-XYZ)
all_skus_to_evaluate = [sku for sublist in storeskus for sku in sublist]

print(f"=== AVALIAÇÃO COMPLETA: {len(all_skus_to_evaluate)} SKUs ===\n")

# Estrutura para armazenar resultados
results_summary = []

# Contadores
successful_predictions = 0
failed_predictions = 0
skus_with_insufficient_data = []

# ===== LOOP POR CADA SKU =====
for idx, sku_to_evaluate in enumerate(all_skus_to_evaluate, 1):
    print(f"\n{'='*100}")
    print(f"[{idx}/{len(all_skus_to_evaluate)}] Avaliando SKU: {sku_to_evaluate}")
    print(f"{'='*100}")
    
    try:
        # ===== 1. CALCULAR BASELINES =====
        df_baseline_calc = df_train[df_train['sku'] == sku_to_evaluate].copy()
        
        if len(df_baseline_calc) < TIMESTEPS + 1:
            print(f" Dados insuficientes no treino: {len(df_baseline_calc)} registros")
            skus_with_insufficient_data.append(sku_to_evaluate)
            failed_predictions += 1
            continue
        
        df_baseline_calc['year'] = df_baseline_calc['date'].dt.year
        df_baseline_calc['month'] = df_baseline_calc['date'].dt.month
        
        # V1: Média por mês do ano
        baseline_by_month_simple = df_baseline_calc.groupby('month')['QTY'].mean().to_dict()
        
        # V2: Mês do ano anterior (2024)
        baseline_by_year_month = df_baseline_calc.groupby(['year', 'month'])['QTY'].mean().reset_index()
        baseline_2024 = baseline_by_year_month[baseline_by_year_month['year'] == 2024].set_index('month')['QTY'].to_dict()
        
        # V3: Moving Average dos últimos 12 meses
        baseline_moving_avg_12m = df_baseline_calc['QTY'].mean()
        
        # ===== 2. PREPARAR DADOS DE TESTE =====
        df_test_only_sku = df_model[
            (df_model['sku'] == sku_to_evaluate) & 
            (df_model['date'] >= '2025-01-01')
        ].copy().reset_index(drop=True)
        
        if len(df_test_only_sku) < TIMESTEPS + 1:
            print(f" Dados insuficientes no teste: {len(df_test_only_sku)} registros")
            skus_with_insufficient_data.append(sku_to_evaluate)
            failed_predictions += 1
            continue
        
        # ===== 3. PREPARAR DADOS PARA MODELO =====
        df_test_model = df_test_only_sku.copy()
        df_test_model.loc[:, available_numeric] = scaler.transform(df_test_model[available_numeric])
        
        # Encoding
        if df_test_model['sku'].iloc[0] in sku_encoder.classes_:
            df_test_model['sku_encoded'] = sku_encoder.transform(df_test_model['sku'])
        else:
            df_test_model['sku_encoded'] = len(sku_encoder.classes_)
        
        if df_test_model['CATEGORY'].iloc[0] in category_encoder.classes_:
            df_test_model['category_encoded'] = category_encoder.transform(df_test_model['CATEGORY'])
        else:
            df_test_model['category_encoded'] = len(category_encoder.classes_)
        
        # Criar sequências
        X_test_numeric, X_test_sku, X_test_category, y_test_real = create_sequences(
            df_test_model, all_features, TIMESTEPS
        )
        
        if len(y_test_real) == 0:
            print(f" Sem sequências suficientes")
            skus_with_insufficient_data.append(sku_to_evaluate)
            failed_predictions += 1
            continue
        
        # ===== 4. PREVISÕES DO MODELO =====
        y_pred_model = model.predict(
            [X_test_numeric, X_test_sku, X_test_category],
            verbose=0
        ).flatten()
        
        # ===== 5. PREVISÕES DOS BASELINES =====
        df_test_sequences = df_test_model.iloc[TIMESTEPS:].reset_index(drop=True)
        test_months = df_test_sequences['month'].values[:len(y_test_real)]
        
        # V1: Média por mês
        y_pred_baseline_v1 = np.array([baseline_by_month_simple.get(m, baseline_moving_avg_12m) for m in test_months])
        
        # V2: Ano anterior
        y_pred_baseline_v2 = np.array([baseline_2024.get(m, baseline_moving_avg_12m) for m in test_months])
        
        # V3: Moving Average
        y_pred_baseline_v3 = np.full(len(y_test_real), baseline_moving_avg_12m)
        
        # ===== 6. CALCULAR MÉTRICAS =====
        from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
        
        def safe_mape(y_true, y_pred):
            return np.mean(np.abs((y_true - y_pred) / np.maximum(y_true, 1))) * 100
        
        # MODELO
        mse_model = mean_squared_error(y_test_real, y_pred_model)
        mae_model = mean_absolute_error(y_test_real, y_pred_model)
        rmse_model = np.sqrt(mse_model)
        r2_model = r2_score(y_test_real, y_pred_model)
        mape_model = safe_mape(y_test_real, y_pred_model)
        
        # BASELINES
        mape_b1 = safe_mape(y_test_real, y_pred_baseline_v1)
        mape_b2 = safe_mape(y_test_real, y_pred_baseline_v2)
        mape_b3 = safe_mape(y_test_real, y_pred_baseline_v3)
        
        # Melhor baseline
        baselines_mape = {'V1': mape_b1, 'V2': mape_b2, 'V3': mape_b3}
        best_baseline_name = min(baselines_mape, key=baselines_mape.get)
        best_baseline_mape = baselines_mape[best_baseline_name]
        
        # Comparar modelo vs melhor baseline
        improvement = ((best_baseline_mape - mape_model) / best_baseline_mape * 100) if best_baseline_mape > 0 else 0
        
        # Categoria ABC-XYZ do SKU
        sku_category = sku_analysis[sku_analysis['sku'] == sku_to_evaluate]['ABC_XYZ'].values
        sku_category = sku_category[0] if len(sku_category) > 0 else 'Unknown'
        
        # Armazenar resultados
        results_summary.append({
            'SKU': sku_to_evaluate,
            'ABC_XYZ': sku_category,
            'N_Test_Months': len(y_test_real),
            'MAPE_Model': mape_model,
            'MAPE_V1': mape_b1,
            'MAPE_V2': mape_b2,
            'MAPE_V3': mape_b3,
            'Best_Baseline': best_baseline_name,
            'Best_Baseline_MAPE': best_baseline_mape,
            'Improvement_%': improvement,
            'MAE_Model': mae_model,
            'RMSE_Model': rmse_model,
            'R2_Model': r2_model,
            'Winner': 'Model' if improvement > 0 else 'Baseline'
        })
        
        # Feedback resumido
        print(f"✓ MAPE Modelo: {mape_model:.2f}% | Melhor Baseline ({best_baseline_name}): {best_baseline_mape:.2f}%")
        print(f"  {' Modelo melhor' if improvement > 0 else '⚠️ Baseline melhor'} ({abs(improvement):.1f}%)")
        
        successful_predictions += 1
        
    except Exception as e:
        print(f"ERRO ao processar {sku_to_evaluate}: {str(e)}")
        failed_predictions += 1
        continue

# ===== 7. RESULTADOS AGREGADOS =====
print("\n" + "="*100)
print("RESUMO GERAL DA AVALIAÇÃO")
print("="*100)
print(f"Total de SKUs avaliados:        {len(all_skus_to_evaluate)}")
print(f"Previsões bem-sucedidas:        {successful_predictions}")
print(f"Falhas (dados insuficientes):   {failed_predictions}")
print(f"\nSKUs com dados insuficientes: {len(skus_with_insufficient_data)}")

if len(results_summary) > 0:
    # Converter para DataFrame
    df_results = pd.DataFrame(results_summary)
    
    print("\n" + "="*100)
    print("ESTATÍSTICAS AGREGADAS")
    print("="*100)
    
    # Média por categoria ABC-XYZ
    print("\n MAPE Médio por Categoria ABC-XYZ:")
    category_stats = df_results.groupby('ABC_XYZ').agg({
        'MAPE_Model': 'mean',
        'Best_Baseline_MAPE': 'mean',
        'Improvement_%': 'mean',
        'SKU': 'count'
    }).round(2)
    category_stats.columns = ['MAPE_Modelo', 'MAPE_Baseline', 'Melhoria_%', 'N_SKUs']
    print(category_stats)
    
    # Performance geral
    print("\n PERFORMANCE GERAL:")
    print(f"MAPE Médio Modelo:            {df_results['MAPE_Model'].mean():.2f}%")
    print(f"MAPE Médio Melhor Baseline:   {df_results['Best_Baseline_MAPE'].mean():.2f}%")
    print(f"Melhoria Média:               {df_results['Improvement_%'].mean():.2f}%")
    
    # Contagem de vitóriFas
    model_wins = (df_results['Winner'] == 'Model').sum()
    baseline_wins = (df_results['Winner'] == 'Baseline').sum()
    
    print(f"\n VITÓRIAS:")
    print(f"Modelo melhor:   {model_wins} SKUs ({model_wins/len(df_results)*100:.1f}%)")
    print(f"Baseline melhor: {baseline_wins} SKUs ({baseline_wins/len(df_results)*100:.1f}%)")
    
    # Distribuição de baselines vencedores
    print("\n Baselines Mais Utilizados:")
    print(df_results['Best_Baseline'].value_counts())
    
    # Top 10 melhores e piores SKUs
    print("\n" + "="*100)
    print("TOP 10 MELHORES PREVISÕES (Menor MAPE Modelo)")
    print("="*100)
    print(df_results.nsmallest(10, 'MAPE_Model')[['SKU', 'ABC_XYZ', 'MAPE_Model', 'Best_Baseline_MAPE', 'Improvement_%']].to_string(index=False))
    
    print("\n" + "="*100)
    print("TOP 10 PIORES PREVISÕES (Maior MAPE Modelo)")
    print("="*100)
    print(df_results.nlargest(10, 'MAPE_Model')[['SKU', 'ABC_XYZ', 'MAPE_Model', 'Best_Baseline_MAPE', 'Improvement_%']].to_string(index=False))
    
    # Salvar resultados
    df_results.to_csv('baseline_comparison_results.csv', index=False)
    print("\n✓ Resultados salvos em: baseline_comparison_results.csv")
    
    # ===== 8. VISUALIZAÇÕES =====
    fig, axes = plt.subplots(2, 2, figsize=(18, 12))
    
    # 1. Comparação MAPE: Modelo vs Baseline
    ax1 = axes[0, 0]
    ax1.scatter(df_results['Best_Baseline_MAPE'], df_results['MAPE_Model'], 
                alpha=0.6, s=50, c=df_results['ABC_XYZ'].astype('category').cat.codes, cmap='viridis')
    max_val = max(df_results['Best_Baseline_MAPE'].max(), df_results['MAPE_Model'].max())
    ax1.plot([0, max_val], [0, max_val], 'r--', linewidth=2, label='Modelo = Baseline')
    ax1.set_xlabel('MAPE Melhor Baseline (%)')
    ax1.set_ylabel('MAPE Modelo LSTM (%)')
    ax1.set_title('Comparação: Modelo vs Melhor Baseline', fontweight='bold')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # 2. Distribuição de Melhoria
    ax2 = axes[0, 1]
    ax2.hist(df_results['Improvement_%'], bins=30, edgecolor='black', alpha=0.7, color='steelblue')
    ax2.axvline(0, color='red', linestyle='--', linewidth=2, label='Sem melhoria')
    ax2.axvline(df_results['Improvement_%'].mean(), color='green', linestyle='--', 
                linewidth=2, label=f'Média: {df_results["Improvement_%"].mean():.1f}%')
    ax2.set_xlabel('Melhoria do Modelo (%)')
    ax2.set_ylabel('Frequência')
    ax2.set_title('Distribuição de Melhoria Relativa', fontweight='bold')
    ax2.legend()
    ax2.grid(True, alpha=0.3, axis='y')
    
    # 3. MAPE por Categoria
    ax3 = axes[1, 0]
    category_mape = df_results.groupby('ABC_XYZ')[['MAPE_Model', 'Best_Baseline_MAPE']].mean()
    x_pos = np.arange(len(category_mape))
    width = 0.35
    ax3.bar(x_pos - width/2, category_mape['MAPE_Model'], width, label='Modelo', alpha=0.8)
    ax3.bar(x_pos + width/2, category_mape['Best_Baseline_MAPE'], width, label='Baseline', alpha=0.8)
    ax3.set_xlabel('Categoria ABC-XYZ')
    ax3.set_ylabel('MAPE Médio (%)')
    ax3.set_title('MAPE Médio por Categoria', fontweight='bold')
    ax3.set_xticks(x_pos)
    ax3.set_xticklabels(category_mape.index, rotation=45)
    ax3.legend()
    ax3.grid(True, alpha=0.3, axis='y')
    
    # 4. Taxa de Sucesso
    ax4 = axes[1, 1]
    success_data = df_results['Winner'].value_counts()
    colors = ['#2ca02c', '#d62728']
    ax4.pie(success_data.values, labels=success_data.index, autopct='%1.1f%%', 
            startangle=90, colors=colors)
    ax4.set_title('Taxa de Sucesso: Modelo vs Baseline', fontweight='bold')
    
    plt.tight_layout()
    plt.show()
    
else:
    print("\n Nenhum resultado válido foi gerado.")
```
