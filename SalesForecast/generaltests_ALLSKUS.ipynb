```python
import pandas as pd

sales = pd.read_csv

# Carregar dados
sales_order = pd.read_csv('../CSVTables/SALES_FLAT_ORDER_CLEAN.csv', encoding='latin-1')
sales_item = pd.read_csv('../CSVTables/SALES_FLAT_ORDER_ITEM_CLEAN.csv', encoding='latin-1')
```

```python
# Verificar se tem tudo
print(f'Sales rows: {sales_order.shape[0]} | Items rows: {sales_item.shape[0]}')

sales_order.head()
```

```python
# Tirar duplicados 
sales_item=sales_item[sales_item['product_type']=='simple']
sales_item.head()
```

```python
# Filtrar successful orders  
sales_order = sales_order[(sales_order['state']=='complete') | (sales_order['state']=='processing')]

# Excluir pedidos com AMZ ou EBAY no increment_id
sales_order = sales_order[~sales_order['increment_id'].str.contains('AMZ|EBAY', na=False)]
```

```python
import matplotlib.pyplot as plt

# Converter created_at para datetime e extrair data
sales_order['created_at'] = pd.to_datetime(sales_order['created_at'])
sales_order['date'] = sales_order['created_at'].dt.date

# Excluir datas depois de abril 2025
sales_order = sales_order[sales_order['created_at'] <= '2025-04-30']

# Agregar revenue por dia
daily_revenue = sales_order.groupby('date')['base_subtotal_with_discount'].sum().reset_index()

# Plotar
plt.figure(figsize=(14, 6))
plt.plot(daily_revenue['date'], daily_revenue['base_subtotal_with_discount'], linewidth=1.5)
plt.xlabel('Data')
plt.ylabel('Revenue (base_subtotal_with_discount)')
plt.title('Revenue Diária')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Agregar custo por dia
daily_cost = sales_order.groupby('date')['base_total_invoiced_cost'].sum().reset_index()

# Criar DataFrame com revenue, custo e lucro
daily_data = daily_revenue.merge(daily_cost, on='date', how='outer')
daily_data = daily_data.fillna(0)
daily_data['profit'] = daily_data['base_subtotal_with_discount'] - daily_data['base_total_invoiced_cost']
daily_data['cumulative_profit'] = daily_data['profit'].cumsum()

# Criar figura com 4 subplots
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. Revenue
axes[0, 0].plot(daily_data['date'], daily_data['base_subtotal_with_discount'], linewidth=1.5, color='green')
axes[0, 0].set_xlabel('Data')
axes[0, 0].set_ylabel('Revenue')
axes[0, 0].set_title('Revenue Diária')
axes[0, 0].tick_params(axis='x', rotation=45)
axes[0, 0].grid(True, alpha=0.3)

# 2. Custo
axes[0, 1].plot(daily_data['date'], daily_data['base_total_invoiced_cost'], linewidth=1.5, color='red')
axes[0, 1].set_xlabel('Data')
axes[0, 1].set_ylabel('Custo')
axes[0, 1].set_title('Custo Diário (base_total_invoiced_cost)')
axes[0, 1].tick_params(axis='x', rotation=45)
axes[0, 1].grid(True, alpha=0.3)

# 3. Lucro
axes[1, 0].plot(daily_data['date'], daily_data['profit'], linewidth=1.5, color='blue')
axes[1, 0].set_xlabel('Data')
axes[1, 0].set_ylabel('Lucro')
axes[1, 0].set_title('Lucro Diário (Revenue - Custo)')
axes[1, 0].tick_params(axis='x', rotation=45)
axes[1, 0].grid(True, alpha=0.3)

# 4. Lucro Acumulado
axes[1, 1].plot(daily_data['date'], daily_data['cumulative_profit'], linewidth=1.5, color='purple')
axes[1, 1].set_xlabel('Data')
axes[1, 1].set_ylabel('Lucro Acumulado')
axes[1, 1].set_title('Lucro Acumulado')
axes[1, 1].tick_params(axis='x', rotation=45)
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Adicionar colunas de ano e mês
daily_data['year'] = pd.to_datetime(daily_data['date']).dt.year
daily_data['month'] = pd.to_datetime(daily_data['date']).dt.to_period('M')

# Preparar dados mensais
# Converter month para datetime no sales_order para agregação
sales_order['month'] = pd.to_datetime(sales_order['date']).dt.to_period('M')

# Agregar revenue e lucro por mês
monthly_data = daily_data.groupby('month').agg({
    'base_subtotal_with_discount': 'sum',
    'profit': 'sum'
}).reset_index()
monthly_data['year'] = monthly_data['month'].dt.year
monthly_data['month_str'] = monthly_data['month'].astype(str)

# Contar orders totais e orders com desconto por mês
monthly_orders_total = sales_order.groupby('month').size().reset_index(name='total_orders')
monthly_discounts = sales_order[sales_order['base_discount_invoiced'] < -0.01].groupby('month').size().reset_index(name='orders_with_discount')

# Contar orders com desconto excluindo cupões (coupon_code == NA ou null)
monthly_discounts_no_coupon = sales_order[
    (sales_order['base_discount_invoiced'] < -0.01) & 
    (sales_order['coupon_code'].isna())
].groupby('month').size().reset_index(name='orders_with_discount_no_coupon')

# Juntar tudo
monthly_data = monthly_data.merge(monthly_orders_total, on='month', how='left').fillna(0)
monthly_data = monthly_data.merge(monthly_discounts, on='month', how='left').fillna(0)
monthly_data = monthly_data.merge(monthly_discounts_no_coupon, on='month', how='left').fillna(0)

# Calcular rácios (percentuais de orders com desconto)
monthly_data['discount_ratio'] = (monthly_data['orders_with_discount'] / monthly_data['total_orders'] * 100).fillna(0)
monthly_data['discount_ratio_no_coupon'] = (monthly_data['orders_with_discount_no_coupon'] / monthly_data['total_orders'] * 100).fillna(0)

# Calcular totais/médias por ano
yearly_totals_revenue = monthly_data.groupby('year')['base_subtotal_with_discount'].sum()
yearly_totals_profit = monthly_data.groupby('year')['profit'].sum()
yearly_avg_discount_ratio = monthly_data.groupby('year')['discount_ratio'].mean()
yearly_avg_discount_ratio_no_coupon = monthly_data.groupby('year')['discount_ratio_no_coupon'].mean()

# Criar figura com 4 subplots (4 linhas, 1 coluna)
fig, axes = plt.subplots(4, 1, figsize=(16, 20))

years = monthly_data['year'].unique()
colors = plt.cm.tab10(range(len(years)))

# ========== GRÁFICO 1: REVENUE ==========
ax_revenue = axes[0]
x_positions = []
for i, year in enumerate(sorted(years)):
    year_data = monthly_data[monthly_data['year'] == year]
    x_pos = range(len(x_positions), len(x_positions) + len(year_data))
    x_positions.extend(x_pos)
    
    bars = ax_revenue.bar(x_pos, year_data['base_subtotal_with_discount'], 
                          color=colors[i], label=f'Ano {int(year)}', alpha=0.8)
    
    # Adicionar label do total do ano
    mid_pos = (x_pos[0] + x_pos[-1]) / 2
    total = yearly_totals_revenue[year]
    ax_revenue.text(mid_pos, ax_revenue.get_ylim()[1] * 0.95, f'Total {int(year)}: {total:,.0f}', 
                   ha='center', va='top', fontsize=11, fontweight='bold', 
                   bbox=dict(boxstyle='round,pad=0.5', facecolor=colors[i], alpha=0.3))

ax_revenue.set_xlabel('Mês')
ax_revenue.set_ylabel('Revenue')
ax_revenue.set_title('Revenue Mensal por Ano')
ax_revenue.set_xticks(range(len(monthly_data)))
ax_revenue.set_xticklabels(monthly_data['month_str'], rotation=45, ha='right')
ax_revenue.grid(True, alpha=0.3, axis='y')
ax_revenue.legend()

# ========== GRÁFICO 2: LUCRO ==========
ax_profit = axes[1]
x_positions = []
for i, year in enumerate(sorted(years)):
    year_data = monthly_data[monthly_data['year'] == year]
    x_pos = range(len(x_positions), len(x_positions) + len(year_data))
    x_positions.extend(x_pos)
    
    bars = ax_profit.bar(x_pos, year_data['profit'], 
                         color=colors[i], label=f'Ano {int(year)}', alpha=0.8)
    
    # Adicionar label do total do ano
    mid_pos = (x_pos[0] + x_pos[-1]) / 2
    total = yearly_totals_profit[year]
    ax_profit.text(mid_pos, ax_profit.get_ylim()[1] * 0.95, f'Total {int(year)}: {total:,.0f}', 
                  ha='center', va='top', fontsize=11, fontweight='bold', 
                  bbox=dict(boxstyle='round,pad=0.5', facecolor=colors[i], alpha=0.3))

ax_profit.set_xlabel('Mês')
ax_profit.set_ylabel('Lucro')
ax_profit.set_title('Lucro Mensal por Ano')
ax_profit.set_xticks(range(len(monthly_data)))
ax_profit.set_xticklabels(monthly_data['month_str'], rotation=45, ha='right')
ax_profit.grid(True, alpha=0.3, axis='y')
ax_profit.legend()

# ========== GRÁFICO 3: RÁCIO DE ORDERS COM DESCONTO ==========
ax_discount = axes[2]
x_positions = []
for i, year in enumerate(sorted(years)):
    year_data = monthly_data[monthly_data['year'] == year]
    x_pos = range(len(x_positions), len(x_positions) + len(year_data))
    x_positions.extend(x_pos)
    
    bars = ax_discount.bar(x_pos, year_data['discount_ratio'], 
                           color=colors[i], label=f'Ano {int(year)}', alpha=0.8)
    
    # Adicionar percentual em cada barra
    for pos, ratio in zip(x_pos, year_data['discount_ratio']):
        ax_discount.text(pos, ratio + 0.5, f'{ratio:.1f}%', 
                        ha='center', va='bottom', fontsize=9)
    
    # Adicionar label da média do ano
    mid_pos = (x_pos[0] + x_pos[-1]) / 2
    avg_ratio = yearly_avg_discount_ratio[year]
    ax_discount.text(mid_pos, ax_discount.get_ylim()[1] * 0.95, f'Média {int(year)}: {avg_ratio:.1f}%', 
                    ha='center', va='top', fontsize=11, fontweight='bold', 
                    bbox=dict(boxstyle='round,pad=0.5', facecolor=colors[i], alpha=0.3))

ax_discount.set_xlabel('Mês')
ax_discount.set_ylabel('% Orders com Desconto')
ax_discount.set_title('Percentual de Orders com Desconto por Mês')
ax_discount.set_xticks(range(len(monthly_data)))
ax_discount.set_xticklabels(monthly_data['month_str'], rotation=45, ha='right')
ax_discount.grid(True, alpha=0.3, axis='y')
ax_discount.legend()

# ========== GRÁFICO 4: RÁCIO DE ORDERS COM DESCONTO (SEM CUPÕES) ==========
ax_discount_no_coupon = axes[3]
x_positions = []
for i, year in enumerate(sorted(years)):
    year_data = monthly_data[monthly_data['year'] == year]
    x_pos = range(len(x_positions), len(x_positions) + len(year_data))
    x_positions.extend(x_pos)
    
    bars = ax_discount_no_coupon.bar(x_pos, year_data['discount_ratio_no_coupon'], 
                                     color=colors[i], label=f'Ano {int(year)}', alpha=0.8)
    
    # Adicionar percentual em cada barra
    for pos, ratio in zip(x_pos, year_data['discount_ratio_no_coupon']):
        ax_discount_no_coupon.text(pos, ratio + 0.5, f'{ratio:.1f}%', 
                                  ha='center', va='bottom', fontsize=9)
    
    # Adicionar label da média do ano
    mid_pos = (x_pos[0] + x_pos[-1]) / 2
    avg_ratio = yearly_avg_discount_ratio_no_coupon[year]
    ax_discount_no_coupon.text(mid_pos, ax_discount_no_coupon.get_ylim()[1] * 0.95, 
                              f'Média {int(year)}: {avg_ratio:.1f}%', 
                              ha='center', va='top', fontsize=11, fontweight='bold', 
                              bbox=dict(boxstyle='round,pad=0.5', facecolor=colors[i], alpha=0.3))

ax_discount_no_coupon.set_xlabel('Mês')
ax_discount_no_coupon.set_ylabel('% Orders com Desconto (Sem Cupões)')
ax_discount_no_coupon.set_title('Percentual de Orders com Desconto por Mês (Excluindo Cupões)')
ax_discount_no_coupon.set_xticks(range(len(monthly_data)))
ax_discount_no_coupon.set_xticklabels(monthly_data['month_str'], rotation=45, ha='right')
ax_discount_no_coupon.grid(True, alpha=0.3, axis='y')
ax_discount_no_coupon.legend()

plt.tight_layout()
plt.show()
```

```python
# Juntar orders com items usando entity_id = order_id
merged_data = sales_item.merge(
    sales_order[['entity_id', 'created_at', 'date']], 
    left_on='order_id', 
    right_on='entity_id', 
    how='inner'
).drop(columns=['entity_id'])  # Remove coluna duplicada

###### ENGANEI ME AQUI - REVENUE ESTAVA ERRADA ANTES NAO É PRECISO * QTY INVOICED AQUI #####
# Calcular revenue correta por SKU: (base_row_total_incl_tax - base_discount_amount) * qty_invoiced
merged_data['sku_revenue'] = (merged_data['base_row_total_incl_tax'] - merged_data['base_discount_amount'])

# Converter date para month (período mensal)
merged_data['month'] = pd.to_datetime(merged_data['date']).dt.to_period('M')

# Agregar por SKU
sku_analysis = merged_data.groupby('sku').agg({
    'sku_revenue': 'sum',    # Revenue total por SKU (usando fórmula correta)
    'qty_invoiced': 'sum'      # Número de vendas
}).reset_index()

sku_analysis.columns = ['sku', 'revenue', 'num_sales']

# Calcular variabilidade (coeficiente de variação) por SKU
# Agrupar vendas por SKU e MÊS (não por dia)
sku_monthly_sales = merged_data.groupby(['sku', 'month'])['qty_invoiced'].sum().reset_index()

# NOVO: Para cada SKU, encontrar mês da primeira e última venda
sku_date_range = sku_monthly_sales.groupby('sku')['month'].agg(['min', 'max']).reset_index()
sku_date_range.columns = ['sku', 'first_sale_month', 'last_sale_month']

print(f"Grid considerando apenas meses entre primeira e última venda de cada SKU...")
print(f"Exemplo de ranges por SKU:")
print(sku_date_range.head(10))

# Para cada SKU, criar range de meses entre primeira e última venda
sku_list = []
for _, row in sku_date_range.iterrows():
    sku = row['sku']
    first_month = row['first_sale_month']
    last_month = row['last_sale_month']
    
    # Criar range de meses para este SKU específico
    sku_months = pd.period_range(start=first_month, end=last_month, freq='M')
    
    # Criar DataFrame com todas as combinações deste SKU
    sku_df = pd.DataFrame({
        'sku': sku,
        'month': sku_months
    })
    
    sku_list.append(sku_df)

# Combinar todos os SKUs
sku_monthly_complete = pd.concat(sku_list, ignore_index=True)

# Merge com os dados reais, preenchendo com 0 onde não há vendas
sku_monthly_complete = sku_monthly_complete.merge(
    sku_monthly_sales, 
    on=['sku', 'month'], 
    how='left'
).fillna(0)

print(f"\nGrid completo criado: {len(sku_monthly_complete)} linhas (SKU x mês relevante)")
print(f"\nExemplo de SKU com meses completos (apenas entre primeira e última venda):")
print(sku_monthly_complete[sku_monthly_complete['sku'] == '000266EV'])

# Calcular estatísticas (mean, std, cv) com os zeros incluídos
sku_std = sku_monthly_complete.groupby('sku')['qty_invoiced'].agg(['mean', 'std']).reset_index()
sku_std['cv'] = (sku_std['std'] / sku_std['mean']).fillna(0)

print(f"\nEstatísticas com zeros incluídos (apenas entre primeira e última venda):")
print(sku_std.head())

# Juntar com análise principal
sku_analysis = sku_analysis.merge(sku_std[['sku', 'cv']], on='sku', how='left')
# print(sku_analysis.head())
# Classificação ABC (baseada em revenue)

sku_analysis = sku_analysis.sort_values('num_sales', ascending=False)
sku_analysis['numsales_cum_pct'] = (sku_analysis['num_sales'].cumsum() / sku_analysis['num_sales'].sum()) * 100
#print(sku_analysis.head())

sku_analysis['ABC'] = 'C'
sku_analysis.loc[sku_analysis['numsales_cum_pct'] <= 80, 'ABC'] = 'A' #  80% rev sao A
sku_analysis.loc[(sku_analysis['numsales_cum_pct'] > 80) & (sku_analysis['numsales_cum_pct'] <= 95), 'ABC'] = 'B' # 80-95% rev sao B
# Classificação XYZ (baseada em coeficiente de variação)
sku_analysis = sku_analysis.sort_values('cv')
sku_analysis['XYZ'] = 'Z'
sku_analysis.loc[sku_analysis['cv'] <= sku_analysis['cv'].quantile(0.33), 'XYZ'] = 'X' # 1º 33% sao X ( interpolacao linear (Max-Min)*0.33 + Min)
sku_analysis.loc[(sku_analysis['cv'] > sku_analysis['cv'].quantile(0.33)) & # 33%-67% sao Y
                 (sku_analysis['cv'] <= sku_analysis['cv'].quantile(0.67)), 'XYZ'] = 'Y'

# Combinar classificações
sku_analysis['ABC_XYZ'] = sku_analysis['ABC'] + sku_analysis['XYZ']

# Mostrar resumo
print(f'\nTotal de SKUs analisados: {len(sku_analysis)}')
print(f'\nDistribuição ABC:')
print(sku_analysis['ABC'].value_counts().sort_index())
print(f'\nDistribuição XYZ:')
print(sku_analysis['XYZ'].value_counts().sort_index())
print(f'\nDistribuição ABC-XYZ:')
print(sku_analysis['ABC_XYZ'].value_counts())

sku_analysis.head(20)
```

```python
# Visualizar matriz ABC-XYZ
abc_xyz_matrix = sku_analysis.groupby(['ABC', 'XYZ']).size().unstack(fill_value=0)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Heatmap da matriz ABC-XYZ
import seaborn as sns
sns.heatmap(abc_xyz_matrix, annot=True, fmt='d', cmap='YlOrRd', ax=axes[0], cbar_kws={'label': 'Número de SKUs'})
axes[0].set_title('Matriz ABC-XYZ - Contagem de SKUs')
axes[0].set_xlabel('Variabilidade (XYZ)')
axes[0].set_ylabel('Volume (ABC)')

# Gráfico de barras por classificação
abc_xyz_counts = sku_analysis['ABC_XYZ'].value_counts().sort_index()
abc_xyz_counts.plot(kind='bar', ax=axes[1], color='steelblue', alpha=0.8)
axes[1].set_title('Distribuição de SKUs por Categoria ABC-XYZ')
axes[1].set_xlabel('Categoria')
axes[1].set_ylabel('Número de SKUs')
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```

```python
# Plotar distribuição do número de vendas por categoria ABC-XYZ
fig, axes = plt.subplots(3, 3, figsize=(18, 14))
fig.suptitle('Distribuição do Número de Vendas por Categoria ABC-XYZ', fontsize=16, y=1.00)

abc_categories = ['A', 'B', 'C']
xyz_categories = ['X', 'Y', 'Z']

for i, abc in enumerate(abc_categories):
    for j, xyz in enumerate(xyz_categories):
        ax = axes[i, j]
        category = abc + xyz
        
        # Filtrar dados para esta categoria
        category_data = sku_analysis[sku_analysis['ABC_XYZ'] == category]['num_sales']
        
        if len(category_data) > 0:
            # Criar histograma
            ax.hist(category_data, bins=30, color='steelblue', alpha=0.7, edgecolor='black')
            ax.set_title(f'{category} (n={len(category_data)})', fontweight='bold')
            ax.set_xlabel('Número de Vendas')
            ax.set_ylabel('Frequência')
            ax.grid(True, alpha=0.3, axis='y')
            
            # Adicionar estatísticas
            mean_val = category_data.mean()
            median_val = category_data.median()
            ax.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Média: {mean_val:.0f}')
            ax.axvline(median_val, color='green', linestyle='--', linewidth=2, label=f'Mediana: {median_val:.0f}')
            ax.legend(fontsize=8)
        else:
            # Se não houver dados, mostrar mensagem
            ax.text(0.5, 0.5, f'{category}\nSem dados', 
                   ha='center', va='center', fontsize=12, transform=ax.transAxes)
            ax.set_xticks([])
            ax.set_yticks([])

plt.tight_layout()
plt.show()
```
