```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path(r"CSVTables")

# lista de ficheiros csv que queres samplear
# -> ajusta com os nomes reais dos teus ficheiros
csv_files = [
    "CATALOG_PRODUCT_ENTITY__VARCHAR_CLEAN.csv",
    "CATALOG_PRODUCT_ENTITY_CLEAN.csv",
    "CATALOG_PRODUCT_ENTITY_INT_CLEAN.csv",
    "EAV_ATTRIBUTE_OPTION_VALUE_CLEAN.csv",
    "SALES_FLAT_ORDER_ADDRESS_CLEAN.csv",
    "SALES_FLAT_ORDER_CLEAN.csv",
    "SALES_FLAT_ORDER_ITEM_CLEAN.csv",
]

# nº de linhas que queres extrair
N_ROWS = 30

def criar_samples(data_dir: Path, files: list[str], n_rows: int = 30):
    for filename in files:
        input_path = data_dir / filename

        if not input_path.exists():
            print(f"[AVISO] Ficheiro não encontrado: {input_path}")
            continue

        try:
            # lê apenas as primeiras n_rows linhas
            df_sample = pd.read_csv(input_path, nrows=n_rows)

            # cria nome para o ficheiro de saída
            output_path = input_path.with_name(input_path.stem + "_sample.csv")

            # guarda o sample
            df_sample.to_csv(output_path, index=False)

            print(f"[OK] Sample criado: {output_path}")

        except Exception as e:
            print(f"[ERRO] Ao ler/gravar {input_path}: {e}")

if __name__ == "__main__":
    criar_samples(DATA_DIR, csv_files, N_ROWS)
```
