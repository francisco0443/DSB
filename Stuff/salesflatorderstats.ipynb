```python
import pandas as pd

# Carregar dados
sales_order = pd.read_csv('CSVTables/SALES_FLAT_ORDER_CLEAN.csv', encoding='latin-1')
sales_item = pd.read_csv('CSVTables/SALES_FLAT_ORDER_ITEM.csv', encoding='latin-1')

# Filtrar apenas pedidos com status == 'closed'
closed_orders = sales_order[sales_order['state'] == 'closed'].copy()

print(f"Total orders with status 'closed': {len(closed_orders)}")

# Agregar qty_backordered por order_id
item_backorder = sales_item.groupby('order_id').agg({
    'qty_backordered': 'sum'
}).reset_index()

# Merge com pedidos closed (relação: ENTITY_ID == ORDER_ID)
result = closed_orders[['entity_id', 'state', 'total_qty_ordered']].merge(
    item_backorder,
    left_on='entity_id',
    right_on='order_id',
    how='left'
)

# Preencher NaN com 0 (pedidos sem backordered)
result['qty_backordered'] = result['qty_backordered'].fillna(0)

# Selecionar colunas finais
result = result[['entity_id', 'state', 'total_qty_ordered', 'qty_backordered']]

print(f"\nTotal rows in result table: {len(result)}")
print(f"\nOrders with backordered items: {len(result[result['qty_backordered'] > 0])}")
    
print(result)
```
